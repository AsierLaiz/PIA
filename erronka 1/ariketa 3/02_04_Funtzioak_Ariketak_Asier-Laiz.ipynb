### 2. Sortu luzera aldakorreko argumentuak dituen funtzio bat.

Idatzi programa bat `func1()` funtzioa sortzeko, luzera aldakorreko argumentu bat onartzeko eta bere balioa inprimatzeko.

Oharra: Sortu edozein argumentu kopuru onar dezakeen, prozesatu eta argumentu bakoitzaren balioa bistaratzen duen funtzio bat.

func1(20, 40, 60)

func1(80, 100)

Emaitza 1:
20
40
60

Emaitza 2:
80
100

```python
def func1(*argumentuak):
    for argumentua in argumentuak:
        print(argumentua)

func1(20, 40, 60)
func1(80, 100)
```

### 4. Sortu funtzio bat argumentu lehenetsi batekin

### 5. Sortu barne-funtzio bat batura kalkulatzeko honela:

Sortu bi parametro onartzen dituen kanpoko funtzio bat, a eta b.
Sortu barne-funtzio bat kanpoko funtzio baten barruan, a eta b-ren batura kalkulatuko duena.
Azkenik, kanpoko funtzio batek 5 batu eta emaitza itzuliko du.

```python
def kan_funtzioa(a, b):
    def bar_funtzioa():
        return a + b
    return bar_funtzioa() + 5
kan_funtzioa(5, 10)
```

### 6. Sortu funtzio errekurtsibo bat
Idatzi programa bat 0tik 10era bitarteko zenbakien batura kalkulatzeko funtzio errekurtsibo bat sortzeko.

Funtzio errekurtsiboa bere burua behin eta berriz deitzen duen funtzioa da.

Irteera: 55

```python
def bat_reb(zen):
    if zen==0:
        return 0
    return zen + bat_reb(zen - 1)
print (bat_reb(10))
```
