### Ariketa 24

Eskatu erabiltzaileari sarrerako testu bat. Begiratu luzera eta adierazi 20 karaktere baino gehiago edo gutxiago dituen. 20 karaktere izanez gero, "PERFECT!" mezua agertuko da:

```python
text = input("Introduce un texto:")
if len(text) < 20:
  print("Es menor de 20 caracteres")
elif len(text) == 20:
  print("PERFECT!")
else:
  print("Es mayor de 20 caracteres")
```

### Ariketa 25

Eskatu erabiltzaileari sarrera-zenbaki bat. Lehenik, egiaztatu zenbaki bat dela; testua bada, "Hau ez da zenbaki bat" mezua agertuko da. Zenbakia bada, egiaztatu 10 baino handiagoa edo txikiagoa den, eta adierazi pantailan:

```python
try:
  num = int(input("Introduce un número:"))
  if num > 10:
    print("El número es mayor que 10")
  elif num == 10:
    print("El número es igual a 10")
  else:
    print("El número es menor que 10")
except:
  print("Esto no es un número")
```

## 3.2 while

### Ariketa 27

Eskatu erabiltzaileari zenbakiak sartzeko eta pantaila bidez erakusteko, 0 bat sartu arte:

```python
num = -1
while num != 0:
  num = int(input("Introduce un número:"))
  print("El número introducido es: ", num)
print("FIN")
```

### Ariketa 28

Eskatu erabiltzaileari zenbakiak sartzeko, guztien baturak 100 zenbakia gainditu arte:

```python
sum = 0
while sum <= 100:
  num = int(input("Introduce un número: "))
  sum += num
  print("La suma actual es: ", sum)
print("FIN")
```

### Ariketa 29

Eskatu erabiltzaileari zenbakiak sartzeko, 0 bat sartu arte. Sartutako zenbakien baturak begiztaren 100 gatz balioa bortxaz gainditzen badu:

```python
sum = 0
num = -1
while num != 0:
  num = int(input("Introduce un número: "))
  sum += num
  print("La suma actual es: ", sum)
  if sum > 100:
    break
print("FIN")
```

### Ariketa 30

Eskatu erabiltzaileari gero eta zenbaki handiagoak sartzeko iterazio bakoitzean. Aurrekoaren berdina edo txikiagoa den zenbaki bat sartzen baduzu, buka ezazu begizta. Gainera, erabiltzaileak beti izango du begiztaren amaiera 0 balioaren bidez behartzeko aukera. Begizta behar bezala amaitu bada (behartu gabe), "END" mezua erakutsiko du:

```python
num = -9999999
old_num = -10000000

while old_num < num:
  old_num = num
  num = int(input("Introduce un nuevo número: "))
  if num == 0:
    break
else:
  print("END")
```

### Ariketa 35

Begiratu "cars" hiztegiko elementu guztiak, eta erakutsi gakoa "S" letrarekin hasten den edo autoa "M" letrarekin hasten den. Begizta bortxaz amaitzen da "C" letrarekin hasten den autoren bat aurkitzen badu:

```python
cars = {
    "Ford": "Fiesta",
    "BMW": "M3",
    "Seat": "Ibiza",
    "Citroen": "C3",
    "Nissan": "Micra"
}

for key, value in cars.items():
  if key[0] == "S":
    print(key)
  if value[0] == "M":
    print(value)
  if value[0] == "C":
    break
```

# 4. Funtzioak

### Ariketa 36

Sarrerako testu bat onartzen duen eta barruan duen bokal kopurua itzultzen duen funtzio bat definitzen du:

```python
def vowel_count(text):
  cont = 0
  for x in text:
    if x == "a" or x == "e" or x == "i" or x == "o" or x == "u" or x == "A" or x == "E" or x == "I" or x == "O" or x == "U":
      cont += 1
  
  return cont

print(vowel_count("Este es un texto de prueba"))
```

### Ariketa 37

Zerrenda bat parametro gisa onartzen duen eta pantailako zerrendan dauden testuetako bokal kopurua erakusten duen funtzio bat definitzen du:

```python
def vowel_count_list(*texts):
  total_vowels = 0
  for x in texts:
    total_vowels += vowel_count(x)
  
  print(total_vowels)

vowel_count_list("Hola", "esto", "es", "una", "prueba")
```

### Ariketa 38

Hiztegi bat sarrera gisa onartzen duen eta bere eduki guztiak pantaila bidez erakusten dituen funtzio bat definitzen du:

```python
def show_dictionary(** my_dictionary):
  for key, value in my_dictionary.items():
    print("[",key,"] - ", value)

show_dictionary(x = "asdf", y = 17, size = "15 cm", weight = 200)
```

### Ariketa 39

Bi zenbakizko parametro onartzen dituen funtzio bat definitzen du, baina, jaso ezean, parametro bakoitzak aurrez definitutako balio bat izatea. Funtzio horrek jasotako bi parametroen batura itzuliko du:

```python
def sum(x = 5, y = 10):
  return x + y

print("sum(3): ", sum(3))
print("sum(x = 3):", sum(x = 3))
print("sum(y = 3):", sum(y = 3))
print("sum(3, 4):", sum(3, 4))
```
