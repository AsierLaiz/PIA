## 1. ariketa
### Idatzi Python programa bat zeroz zatitzean ZeroDivisionError salbuespen bat kudeatzeko.

```python
def division(a,b):
    try:
        emaitza = a/b
        print( emaitza)
    except ZeroDivisionError:
        print("Errorea: Ezin da zeroz zatitu")

division (10,2)
division (0,5)
division (5,0)
```

## 2. ariketa
### Idatzi Python programa bat, erabiltzaileari zenbaki oso bat sartzeko eskatzen diona eta sarrera zenbaki oso balioduna ez bada ValueError salbuespena botatzen duena.

```python
def zen_oso():
    try:
        sarrera=input("sartu zenbaki bat")
        zenbakia= int(sarrera)
        print(f"ondo {zenbakia}zenbakia da")
    except ValueError:
        print("zenbaki bat izan behar da")

zen_oso()

```

## 3. ariketa
### Idatzi Python programa bat, fitxategi bat ireki eta FileNotFoundError salbuespena kudeatzen duena fitxategia existitzen ez bada.

```python
def ireki_fitx():
    sarrera=input("Fitxategiaren izena")

    try:
        with open(sarrera,"r",encoding="utf-8") as fitxategia:
            edukia = fitxategia.read()
            print(edukia)
    except FileNotFoundError:
        print("Ezin da fitxategia aurkitu")

ireki_fitx()
```

## 4. ariketa
### Idatzi Python programa bat, erabiltzaileari bi zenbaki sartzeko eskatzen diona eta sarrerak ez badira zenbakizkoak TypeError salbuespena botatzen duena.

```python

def zenbaki():
    zenbaki1=input("sartu zenbaki")
    zenbaki2=input("beste bat")
    try:
        zenbakia= float(zenbaki1)
        zenbakia2= float(zenbaki2)
    except ValueError:
        raise TypeError("zenbaki bat sartu")
    print(f"zure zenbakiak {zenbakia} eta {zenbakia2}")


try:
    zenbaki()
except TypeError as e:
    print(e)
```

## 5. ariketa

#### Idatzi Python programa bat, fitxategi bat ireki eta baimen arazoren bat badago PermissionError salbuespen bat kudeatzen duena.

```python

def ireki_fitx_perm():
    sarrera=input("Fitxategiaren izena")

    try:
        with open(sarrera,"r",encoding="utf-8") as fitxategia:
            edukia = fitxategia.read()
            print(edukia)
    except PermissionError:
        print("Ezin da fitxategia aurkitu")


ireki_fitx_perm()
```
