## Ejercicio 1: Tu primer módulo: texto.py

Crea un módulo `texto.py` con estas funciones (cada una con un docstring corto):

- `contar_vocales(frase)`: devuelve cuántas vocales tiene la frase (sin distinguir mayúsculas).
- `es_palindromo(palabra)`: devuelve `True` si la palabra se lee igual al revés (sin distinguir mayúsculas), `False` si no.
- `invertir(frase)`: devuelve la frase dada la vuelta.

En otra celda, importa el módulo con `import texto` y prueba las tres funciones al menos con un valor cada una.

````python
%%writefile texto.py
def contar_vocales(frase):
    """Devuelve el número de vocales de la frase."""
    return len([c for c in frase.lower() if c in "aeiou"])


def es_palindromo(palabra):
    """Devuelve True si la palabra se lee igual al revés."""
    palabra = palabra.lower()
    return palabra == palabra[::-1]


def invertir(frase):
    """Devuelve la frase dada la vuelta."""
    return frase[::-1]
````

````python
import texto

print(texto.contar_vocales("Murcielago"))
print(texto.es_palindromo("Ana"))
print(texto.invertir("kaixo"))
````

## Ejercicio 4: Módulo con errores controlados: descuento.py

Crea un módulo `descuento.py` con una función `aplicar_descuento(precio, porcentaje)`:

- Si `precio <= 0`, devuelve el texto `"Error: precio no válido"`.
- Si `porcentaje < 0` o `porcentaje > 100`, devuelve el texto `"Error: porcentaje no válido"`.
- En caso contrario, devuelve el precio final después de aplicar el descuento.

Importa el módulo y prueba la función con 3 casos: uno correcto y dos que provoquen cada error.

````python
%%writefile descuento.py
def aplicar_descuento(precio, porcentaje):
    """Devuelve el precio con descuento o un mensaje de error."""
    if precio <= 0:
        return "Error: precio no válido"
    if porcentaje < 0 or porcentaje > 100:
        return "Error: porcentaje no válido"
    return precio - precio * porcentaje / 100
````

````python
import descuento

print(descuento.aplicar_descuento(100, 20))
print(descuento.aplicar_descuento(-5, 10))
print(descuento.aplicar_descuento(50, 150))
````

## Ejercicio 6: Mini-proyecto: boletín de notas

Crea un mini-proyecto con tres ficheros:

- `alumnos.py`: un diccionario `notas = {"Ane": 8, "Jon": 4, "Leire": 6}` y una función `nota(nombre)` que devuelva la nota o el texto `"Alumno no encontrado"` si no existe.
- `boletin.py`: importa `alumnos` y define `media(lista_nombres)`, que devuelve la media de las notas de los alumnos de la lista usando `alumnos.nota()`. Los alumnos que no existan se ignoran. Si no queda ninguno, devuelve 0.
- `main.py`: importa `boletin`, define una función `main()` que muestre la media de `["Ane", "Jon", "Pepe", "Leire"]`, y la llama solo si el fichero se ejecuta directamente (`if __name__ == "__main__":`).

Ejecuta `main.py` desde el notebook con `%run main.py`. Todas las funciones con docstring.

````python
%%writefile alumnos.py
notas = {"Ane": 8, "Jon": 4, "Leire": 6}


def nota(nombre):
    """Devuelve la nota del alumno o un mensaje si no existe."""
    return notas.get(nombre, "Alumno no encontrado")
````

````python
%%writefile boletin.py
import alumnos


def media(lista_nombres):
    """Media de las notas de los alumnos que existan. Ignora los demás."""
    total = 0
    cantidad = 0
    for nombre in lista_nombres:
        valor = alumnos.nota(nombre)
        if valor != "Alumno no encontrado":
            total += valor
            cantidad += 1
    if cantidad == 0:
        return 0
    return total / cantidad
````

````python
%%writefile main.py
import boletin


def main():
    """Muestra la media de una lista de alumnos."""
    print("Media:", boletin.media(["Ane", "Jon", "Pepe", "Leire"]))


if __name__ == "__main__":
    main()
````

````python
%run main.py
````

## Ejercicio 7: Módulo de utilidades con ficheros

a) Crea un módulo `utilidades.py` con dos funciones:

- `leer_lineas(ruta)`: devuelve una lista con las líneas del fichero sin el salto de línea. Si el fichero no existe, captura el `FileNotFoundError` y devuelve una lista vacía.
- `guardar_lineas(ruta, lineas)`: escribe cada elemento de la lista en una línea del fichero (sobrescribiendo lo que hubiera).

b) Importa las dos funciones con `from utilidades import ...`.
c) Guarda `["Ane", "Jon", "Leire"]` en `nombres.txt` y léelo de nuevo mostrando el resultado.
d) Lee un fichero que no existe y comprueba que se obtiene una lista vacía, sin error.
e) Lee `nombres.txt`, pasa cada nombre a mayúsculas y guárdalos en `mayusculas.txt`. Léelo y muéstralo.

````python
%%writefile utilidades.py
def leer_lineas(ruta):
    """Devuelve las líneas del fichero sin salto de línea ([] si no existe)."""
    try:
        with open(ruta, "r", encoding="utf-8") as f:
            return [linea.strip() for linea in f]
    except FileNotFoundError:
        return []


def guardar_lineas(ruta, lineas):
    """Escribe cada elemento de la lista en una línea del fichero."""
    with open(ruta, "w", encoding="utf-8") as f:
        for linea in lineas:
            f.write(linea + "\n")
````

````python
from utilidades import leer_lineas, guardar_lineas

guardar_lineas("nombres.txt", ["Ane", "Jon", "Leire"])
print(leer_lineas("nombres.txt"))

print(leer_lineas("no_existe.txt"))

nombres = leer_lineas("nombres.txt")
guardar_lineas("mayusculas.txt", [n.upper() for n in nombres])
print(leer_lineas("mayusculas.txt"))
````

## Ejercicio 8: Estilo examen: paquete biblioteca

Crea un paquete `biblioteca` con esta estructura:

```
biblioteca/
    __init__.py
    libros.py
    estadisticas.py
main.py
```

- `libros.py`: un diccionario `catalogo = {"Hamlet": 2, "Dune": 0, "Ulises": 1}` (título: copias disponibles) y dos funciones:
  - `prestar(titulo)`: si el libro no existe devuelve `"Error: libro no encontrado"`; si no quedan copias devuelve `"Error: sin copias disponibles"`; si no, resta una copia y devuelve las copias que quedan.
  - `devolver(titulo)`: si el libro no existe devuelve `"Error: libro no encontrado"`; si no, suma una copia y devuelve las copias que hay ahora.
- `estadisticas.py`: importa `libros` desde el paquete y define `total_copias()` (suma de todas las copias) y `agotados()` (lista de títulos con 0 copias).
- `main.py`: con `from biblioteca import libros, estadisticas`, una función `main()` que haga, mostrando cada resultado: prestar "Hamlet", prestar "Dune", prestar "Quijote", devolver "Dune", mostrar el total de copias y mostrar la lista de agotados. Solo se ejecuta si el fichero se lanza directamente.

Todas las funciones con docstring. Ejecuta con `%run main.py`.

````python
import os
os.makedirs("biblioteca", exist_ok=True)
````

````python
%%writefile biblioteca/__init__.py
````

````python
%%writefile biblioteca/libros.py
catalogo = {"Hamlet": 2, "Dune": 0, "Ulises": 1}


def prestar(titulo):
    """Presta un libro: devuelve las copias que quedan o un mensaje de error."""
    if titulo not in catalogo:
        return "Error: libro no encontrado"
    if catalogo[titulo] == 0:
        return "Error: sin copias disponibles"
    catalogo[titulo] -= 1
    return catalogo[titulo]


def devolver(titulo):
    """Devuelve un libro: devuelve las copias que hay ahora o un mensaje de error."""
    if titulo not in catalogo:
        return "Error: libro no encontrado"
    catalogo[titulo] += 1
    return catalogo[titulo]
````

````python
%%writefile biblioteca/estadisticas.py
from biblioteca import libros


def total_copias():
    """Suma de todas las copias del catálogo."""
    return sum(libros.catalogo.values())


def agotados():
    """Lista de títulos sin copias."""
    return [titulo for titulo, copias in libros.catalogo.items() if copias == 0]
````

````python
%%writefile main.py
from biblioteca import libros, estadisticas


def main():
    """Simula varias operaciones de la biblioteca."""
    print(libros.prestar("Hamlet"))
    print(libros.prestar("Dune"))
    print(libros.prestar("Quijote"))
    print(libros.devolver("Dune"))
    print(estadisticas.total_copias())
    print(estadisticas.agotados())


if __name__ == "__main__":
    main()
````
