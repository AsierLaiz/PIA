## Ejercicio 1

Pide dos números y divídelos. Controla `ValueError` (texto en vez de número) y `ZeroDivisionError`, con un mensaje distinto para cada uno.

```python
try:
    a = float(input("Primer número: "))
    b = float(input("Segundo número: "))
    print("Resultado:", a / b)
except ValueError:
    print("Error: debes introducir números")
except ZeroDivisionError:
    print("Error: no se puede dividir entre 0")
```

## Ejercicio 2

Dada `frutas = ["sagarra", "gozokia", "platanoa"]`, pide un índice y muestra el elemento. Controla `IndexError` y `ValueError`.

```python
frutas = ["sagarra", "gozokia", "platanoa"]

try:
    indice = int(input("Índice: "))
    print(frutas[indice])
except IndexError:
    print("Error: el índice está fuera de rango")
except ValueError:
    print("Error: debes introducir un número entero")
```

## Ejercicio 3

Dado `d = {"a": 1, "b": 2}`, pide una clave y muestra su valor. Si no existe, captura `KeyError` y avisa al usuario.

```python
d = {"a": 1, "b": 2}

clave = input("Clave: ")
try:
    print(d[clave])
except KeyError:
    print(f"Error: la clave '{clave}' no existe")
```

## Ejercicio 4

Crea `dividir(a, b)` que lance con `raise` un `ValueError("b no puede ser 0")` si `b == 0`. Llámala dentro de un `try/except` y muestra el mensaje del error.

```python
def dividir(a, b):
    if b == 0:
        raise ValueError("b no puede ser 0")
    return a / b

try:
    print(dividir(10, 0))
except ValueError as e:
    print("Error:", e)
```

## Ejercicio 5

Pide el nombre de un fichero y muestra su contenido. Controla `FileNotFoundError` y añade un `finally` que muestre "Fin del programa".

```python
nombre = input("Nombre del fichero: ")

try:
    with open(nombre, "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print(f"Error: el fichero '{nombre}' no existe")
finally:
    print("Fin del programa")
```
