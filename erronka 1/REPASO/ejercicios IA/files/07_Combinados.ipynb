## Ejercicio 1

Pide al usuario una lista de notas separadas por comas (ej.: `7, 4.5, 9`), conviértelas a números (controlando errores con `try/except`) y muestra la media, la nota mayor y cuántas aprueban (>= 5).

```python
lista= input("sartu lista")


try:
    notas = [float(n) for n in lista.split(",")]    
except ValueError:
    print(",")    
else:
    media=sum(notas)/len(notas)
    aprobado=len([n for n in notas if n >= 5])
    print(media)
    print(aprobado)
    print(max(notas))
```

## Ejercicio 2

Crea una función que reciba un diccionario `{producto: precio}` y un nombre de producto, y devuelva el precio. Si no existe, que devuelva un mensaje de error usando `try/except`.

```python
def dic(productos, nombre):
    try:
        return productos[nombre]
    except KeyError:
        print("aaaa")



tienda = {"apple": 0.5, "bread": 1.2, "milk": 0.9}

print(dic(tienda, "bread"))
print(dic(tienda, "wine"))

```

## Ejercicio 3

Lee un fichero de texto, guarda cada línea en una lista y cuenta cuántas líneas tienen más de 20 caracteres. Controla `FileNotFoundError`.

```python
try:
    with open("file.text","r",encoding="utf-8") as f:
        lineas = [linea.strip() for linea in f]
    largas = len([l for l in lineas if len(l) > 20])
    print("Líneas con más de 20 caracteres:", largas)
except FileNotFoundError:
    print("no existe")
```

## Ejercicio 4

Dada una lista de tuplas `[("Ane", 8), ("Jon", 4), ("Leire", 6)]` (nombre, nota), crea una función que devuelva un diccionario solo con los aprobados y otra que devuelva el nombre del alumno con mejor nota.

```python
lista = [("Ane", 8), ("Jon", 4), ("Leire", 6)]
def apr(list):
    return {nombre: nota for nombre, nota in list if nota >=5 }
def mej(list):
    return max(lista, key=lambda t: t[1])[0]

print(apr(lista))
print(mej(lista))
```

## Ejercicio 5

Pide nombres de alumnos y sus notas en un bucle hasta que el usuario escriba `fin`. Guarda los datos en un diccionario (controlando que la nota sea un número) y al final muestra la media y quién tiene la nota más alta.

```python
notas = {}

while True:
    nombre = input("Nombre (fin para terminar): ")
    if nombre == "fin":
        break
    try:
        notas[nombre] = float(input("Nota: "))
    except ValueError:
        print("Error: la nota debe ser un número")

if notas:
    media = sum(notas.values()) / len(notas)
    mejor = max(notas, key=notas.get)
    print("Media:", media)
    print("Nota más alta:", mejor, notas[mejor])
```
