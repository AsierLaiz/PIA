# Diccionarios - Soluciones

## Ejercicio 1

Dadas `productos = ["pan", "leche", "queso"]` y `precios = [1.2, 0.9, 6.5]`, crea un diccionario y muestra solo los productos que cuestan más de 5 €.

```python
productos = ["pan", "leche", "queso"]
precios = [1.2, 0.9, 6.5]

d = dict(zip(productos, precios))
caros = {p: pr for p, pr in d.items() if pr > 5}
print(caros)
```

## Ejercicio 2

Pide una palabra al usuario y cuenta cuántas veces aparece cada letra, guardando el resultado en un diccionario.

```python
palabra = input("Palabra: ")

cuenta = {}
for letra in palabra:
    cuenta[letra] = cuenta.get(letra, 0) + 1
print(cuenta)
```

## Ejercicio 3

Dado `alumnos = {"Ane": 8, "Koldo": 4, "Leire": 6}`, muestra el alumno con la mejor nota y crea un diccionario nuevo solo con los aprobados (nota >= 5).

```python
alumnos = {"Ane": 8, "Koldo": 4, "Leire": 6}

mejor = max(alumnos, key=alumnos.get)
print("Mejor nota:", mejor, alumnos[mejor])

aprobados = {nombre: nota for nombre, nota in alumnos.items() if nota >= 5}
print(aprobados)
```

## Ejercicio 4

Dado `emp = {"e1": {"nombre": "Jon", "sueldo": 1500}, "e2": {"nombre": "Emma", "sueldo": 1800}}`, sube el sueldo un 10 % a todos y muestra el total de sueldos.

```python
emp = {"e1": {"nombre": "Jon", "sueldo": 1500}, "e2": {"nombre": "Emma", "sueldo": 1800}}

total = 0
for clave in emp:
    emp[clave]["sueldo"] = round(emp[clave]["sueldo"] * 1.1, 2)
    total += emp[clave]["sueldo"]

print(emp)
print("Total:", total)
```

## Ejercicio 5

Dado `d = {"name": "Kelly", "city": "Bilbao"}`, cambia el nombre de la clave `city` por `location`, comprobando antes que la clave existe.

```python
d = {"name": "Kelly", "city": "Bilbao"}

if "city" in d:
    d["location"] = d.pop("city")
print(d)
```
