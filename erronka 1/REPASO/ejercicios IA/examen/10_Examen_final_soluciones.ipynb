# Ejercicio final tipo examen: Inventario de una tienda - Solución

## Módulo `tienda.py` y fichero de datos

```python
%%writefile tienda.py
def leer_productos(ruta):
    """Lee el CSV y devuelve una lista de tuplas (nombre, precio, stock)."""
    productos = []
    try:
        with open(ruta, "r", encoding="utf-8") as f:
            for linea in f:
                campos = linea.strip().split(";")
                try:
                    productos.append((campos[0], float(campos[1]), int(campos[2])))
                except ValueError:
                    print("Línea con error:", linea.strip())
    except FileNotFoundError:
        print("El fichero", ruta, "no existe")
        return []
    return productos


def valor_inventario(lista):
    """Devuelve el valor total del inventario (precio x stock)."""
    total = 0
    for producto in lista:
        total += producto[1] * producto[2]
    return total
```

## a) Leer el fichero

```python
from tienda import leer_productos, valor_inventario

# a) leer el fichero (las líneas con error se avisan y se saltan)
lista = leer_productos("productos.csv")
print(lista)
```

## b) Diccionario

```python
# b) lista de tuplas -> diccionario {nombre: (precio, stock)}
inventario = {nombre: (precio, stock) for nombre, precio, stock in lista}
print(inventario)
```

## c) Valor del inventario

```python
# c) valor total del inventario
print("Valor del inventario:", valor_inventario(lista))
```

## d) Sin stock y más caro

```python
# d) productos sin stock y producto más caro
sin_stock = [p[0] for p in lista if p[2] == 0]
print("Sin stock:", sin_stock)

mas_caro = max(lista, key=lambda p: p[1])
print("Más caro:", mas_caro[0], mas_caro[1])
```

## e) Menú

```python
# e) menú con while
while True:
    print("1. Consultar precio")
    print("2. Vender")
    print("3. Salir")
    opcion = input("Opción: ")

    if opcion == "1":
        nombre = input("Producto: ")
        try:
            print("Precio:", inventario[nombre][0])
        except KeyError:
            print("Ese producto no existe")

    elif opcion == "2":
        nombre = input("Producto: ")
        try:
            cantidad = int(input("Cantidad: "))
            precio, stock = inventario[nombre]
            if cantidad <= 0:
                print("Error: la cantidad debe ser mayor que 0")
            elif cantidad > stock:
                print("Error: stock insuficiente")
            else:
                # las tuplas no se modifican: se crea una nueva
                inventario[nombre] = (precio, stock - cantidad)
                print("Venta realizada")
        except ValueError:
            print("Error: la cantidad debe ser un número entero")
        except KeyError:
            print("Ese producto no existe")

    elif opcion == "3":
        break
    else:
        print("Opción no válida")
```

## f) Fichero final

```python
# f) escribir el stock final
with open("stock_final.txt", "w", encoding="utf-8") as f:
    for nombre, (precio, stock) in inventario.items():
        f.write(f"{nombre}: {stock}\n")
```
