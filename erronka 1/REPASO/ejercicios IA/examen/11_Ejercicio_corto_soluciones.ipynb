# Ejercicio corto: Notas de una clase - Solución

## a) Función media

```python
notas = {"Ane": 7, "Jon": 4, "Leire": 9, "Mikel": 5}

def media(d):
    """Devuelve la nota media del diccionario."""
    total = 0
    for nota in d.values():
        total += nota
    return total / len(d)

print("Media:", media(notas))
```

## b) Aprobados

```python
aprobados = [alumno for alumno, nota in notas.items() if nota >= 5]
print("Aprobados:", aprobados)
```

## c) Bucle para actualizar notas

```python
while True:
    nombre = input("Nombre (fin para terminar): ")
    if nombre == "fin":
        break

    try:
        print("Nota actual:", notas[nombre])
    except KeyError:
        print("Ese alumno no existe")
        continue          # vuelve a pedir el nombre

    try:
        nueva = float(input("Nueva nota: "))
        notas[nombre] = nueva
    except ValueError:
        print("Error: la nota debe ser un número")
```

## d) Fichero final y nueva media

```python
with open("notas_final.txt", "w", encoding="utf-8") as f:
    for alumno, nota in notas.items():
        f.write(f"{alumno}: {nota}\n")

print("Nueva media:", media(notas))
```
