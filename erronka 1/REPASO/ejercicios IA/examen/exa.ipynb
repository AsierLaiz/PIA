```python
import tienda

contenido=tienda.leer_productos("productos.csv")
print(contenido)


hiztegi={nombre: (precio, stock) for nombre, precio, stock in contenido}
print(hiztegi)


print("Valor del inventario:", tienda.valor_inventario(contenido))


nostock=[p[0] for p in contenido if p[2]==0]
print(nostock)

preciomax=max(contenido,key=lambda e:e[1])
print(preciomax)

seguir = True
while seguir:
    respuesta = int(input("Selecciona 1 para consultar precio, 2 para vender y tres para salir bobaina"))
    match respuesta: 
        case 1:
            seleccionar=int(input("mete producto"))
            try:
                tienda.consultarPrecio(contenido[seleccionar])
            except IndexError: 
                print("No existeeeee")
        case 2:
            seleccionar2=int(input("mete producto"))
            cantidat=int(input("mete ya"))
            precio,stock=hiztegi["nombre"]
            tienda.vender(seleccionar2,cantidat,stock,hiztegi)
        case 3:
            seguir = False
        case _:
            respuesta
```
