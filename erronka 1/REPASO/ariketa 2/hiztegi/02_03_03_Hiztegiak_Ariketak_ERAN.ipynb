### 1. Bi zerrenda hiztegi bihurtzea

Bi zerrenda emanda, idatzi Python programa bat hiztegi bihurtzeko, non list1-en elementua gakoa den eta list2-en elementua balioa.

keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]

Emaitza: {'Ten': 10, 'Twenty': 20, 'Thirty': 30}

```python
# Solución 1 usando función zip()
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]

res_dict = dict(zip(keys, values))
print(res_dict)

# Solución 2
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]

# empty dictionary
res_dict = dict()

for i in range(len(keys)):
    res_dict.update({keys[i]: values[i]})
print(res_dict)
```

### 2. Bi Python hiztegi batu

dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = {'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

{'Ten': 10, 'Twenty': 20, 'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

```python
'''Solución 1:
    Un asterisco (*): Se usa para desempaquetar elementos de una lista o cualquier iterable.
    Dos asteriscos (**): Se usan para desempaquetar elementos de un diccionario. Cuando usamos la sintaxis {**dict1, **dict2}, estamos 
    desempaquetando los pares clave-valor de dict1 y dict2 dentro de un nuevo diccionario.
'''

dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = {'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

dict3 = {**dict1, **dict2}
print(dict3)

dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = {'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

dict3 = dict1.copy()
dict3.update(dict2)
print(dict3)

```

### 4. Hasieratu hiztegia balio lehenetsiekin. Inprimatu 'Kelly' gakoaren balioa.

Erabili `fromkeys()` funtzioa

employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}

Emaitza: {'Kelly': {'designation': 'Developer', 'salary': 8000}, 'Emma': {'designation': 'Developer', 'salary': 8000}}


```python
employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}

res = dict.fromkeys(employees, defaults)
print(res)


# Individual data
print(res["Kelly"])

res["Kelly"]["salary"] = 9000
print(res["Kelly"]["salary"])  
print(res["Emma"]["salary"])  


#res = {emp: defaults.copy() for emp in employees}
#print(res)
```

### 5. Sortu hiztegi bat hiztegi jakin batetik gakoak ateraz

Idatzi Python programa bat hiztegi berri bat sortzeko, zehaztutako gakoak hiztegi honetatik ateraz.

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"}

keys = ["name", "salary"]

```python
# Solución 1: iterando las claves y método update

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"}

# keys to extract
keys = ["name", "salary"]

# new dict
res = dict()

for k in keys:
    # add current key with its value from sample_dict
    res.update({k: sample_dict[k]})
print(res) 

# Solución 2: Comprensión de diccionarios

sampleDict = { 
  "name": "Kelly",
  "age":25, 
  "salary": 8000, 
  "city": "New york" }

keys = ["name", "salary"]

newDict = {k: sampleDict[k] for k in keys}
print(newDict)
```

### 6. Ezabatu hiztegi bateko gakoen zerrenda

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

#Keys to remove
keys = ["name", "salary"]

Erabili `pop()` metodoa

```python
# Solución 1:  usando método pop()

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}
# Keys to remove
keys = ["name", "salary"]

for k in keys:
    sample_dict.pop(k)
print(sample_dict)

# solución 2: comprensión de diccionarios
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}
# Keys to remove
keys = ["name", "salary"]

sample_dict = {k: sample_dict[k] for k in sample_dict.keys() - keys}
print(sample_dict)
```
