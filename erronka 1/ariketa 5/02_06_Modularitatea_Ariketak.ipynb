## 1. ariketa — Zure lehen modulu pertsonalizatua

Arazoa: Sortu convertor.py izeneko fitxategi bat, funtzio hauek dituena:

- celsius_to_fahrenheit(c) → tenperatura Fahrenheit-etan itzultzen du
- fahrenheit_to_celsius(f) → tenperatura Celsius-etan itzultzen du
- km_to_miles(km) → distantzia milietan itzultzen du

Ondoren, beste fitxategi (edo gelaxka) batean, inportatu bihurgailu modulua eta probatu hiru funtzioak gutxienez balio bana dutela.

# 2. ariketa — Modulua inportatzea VS Kodea Exekutazea

Aldatu aurreko ariketako converter.py modulua fitxategiaren amaieran proba-lerro bat gehituz: `print("Converter modulua kargatuta").`

Egiaztatu zer gertatzen den bi aldiz inportatzen duzunean (adibidez, bi gelaxka desberdinetatik) eta azaldu iruzkin batean zergatik `print` adierazpen hau lehenengo aldian bakarrik exekutatzen den.

Ondoren, zuzendu modulua `if __name__ == "__main__":` eraikuntza erabiliz, mezu hau fitxategia zuzenean exekutatzen bada bakarrik bistara dadin, eta ez inportatzen denean.

# 3. ariketa — Erroreen kudeaketa duen modulua

bank.py izeneko modulua sortu erretiratu(saldoa, zenbatekoa) funtzio batekin:

- Saldo berria itzultzen du zenbatekoa <= saldoa eta zenbatekoa > 0 bada
- "Errorea: Saldo nahikorik ez" katea itzultzen du zenbatekoa > saldoa bada
- "Errorea: Zenbateko baliogabea" katea itzultzen du zenbatekoa <= 0 bada

Inportatu modulua eta probatu funtzioa gutxienez 3 kasu ezberdinekin (bat arrakastatsua eta bi erroreak sortzen dituztenak).

# 4. ariketa — Zure paketea sortzea

Sortu figures izeneko pakete bat egitura honekin:

figures/

    __init__.py
    circle.py
    rectangulu.py

circle.py-k area(radius) funtzio bat eta beste perimetro(radius) bat izan behar ditu.
rectangle.py-k area(base, height) funtzio bat eta beste perimetro(base, height) bat izan behar ditu.
__init__.py hutsik utz daiteke.

Beste fitxategi batetik, inportatu bi moduluak paketetik (from figures import circle, rectangle) eta kalkulatu 5 erradioko zirkulu baten eta 4 oinarri eta 6 altuerako laukizuzen baten azalera eta perimetroa.

# 5. ariketa — Mini-proiektu modularra

Eraiki "denda kalkulagailu" txiki bat modulutan banatuta:

`inventory.py:` produktu hiztegi bat ({"apple": 0.5, "bread": 1.2, "milk": 0.9}) eta prezioa edo "Produktua ez da aurkitu" itzultzen duen price(product) funtzio bat.
`cart.py:` produktu izenen zerrenda bat jasotzen duen calculate_total(product_list) funtzio bat, bakoitzerako inventory.price() erabiltzen duena eta ordaindu beharreko zenbateko osoa itzultzen duena. Produkturen bat existitzen ez bada, alde batera utzi (edo abisatu) beharko luke.

Fitxategi nagusi bat, `main.py`, cart inportatzen duena eta ["apple", "bread", "milk", "apple"] erosketa simulatzen duena, azken guztizkoa erakutsiz.

`Baldintza gehigarriak:`

- Erabili if __name__ == "__main__": main.py-n.

- Dokumentatu funtzio bakoitza docstring labur batekin.
