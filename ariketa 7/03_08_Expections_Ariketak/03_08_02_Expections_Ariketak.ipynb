## 1. ariketa
### Idatzi Python programa bat zeroz zatitzean ZeroDivisionError salbuespen bat kudeatzeko.

```python
def zatitu(a, b):
    try:
        emaitza = a / b
        print(f"{a} / {b} = {emaitza}")
    except ZeroDivisionError:
        print(f"Errorea: Ezin da {a} zenbakia zeroz zatitu.")


zatitu(10, 2)

zatitu(8, 0)
```

## 2. ariketa
### Idatzi Python programa bat, erabiltzaileari zenbaki oso bat sartzeko eskatzen diona eta sarrera zenbaki oso balioduna ez bada ValueError salbuespena botatzen duena.

```python

def eskatu_zenbaki_osoa():
    try:
        sarrera = input("Sartu zenbaki oso bat: ")
        zenbakia = int(sarrera)
        print(f"Ederki! Sartutako zenbaki osoa {zenbakia} da.")
        
    except ValueError:
        print("Errorea: Sartutako testua ez da zenbaki oso balioduna.")

eskatu_zenbaki_osoa()
```

## 3. ariketa
### Idatzi Python programa bat, fitxategi bat ireki eta FileNotFoundError salbuespena kudeatzen duena fitxategia existitzen ez bada.

```python
def irakurri_fitxategia():
    fitxategi_izena = input("Sartu ireki nahi duzun fitxategiaren izena: ")

    try:
        with open(fitxategi_izena, "r", encoding="utf-8") as fitxategia:
            edukia = fitxategia.read()
            print(f"\n--- '{fitxategi_izena}' fitxategiaren edukia ---")
            print(edukia)

    except FileNotFoundError:
        print(f"\nErrorea: Ezin izan da '{fitxategi_izena}' fitxategia aurkitu. Ziurtatu izena ondo idatzita dagoela.")

irakurri_fitxategia()
```

## 4. ariketa
### Idatzi Python programa bat, erabiltzaileari bi zenbaki sartzeko eskatzen diona eta sarrerak ez badira zenbakizkoak TypeError salbuespena botatzen duena.

```python
def batu_zenbakiak():
    sarrera1 = input("Sartu lehen zenbakia: ")
    sarrera2 = input("Sartu bigarren zenbakia: ")

    try:
        z1 = float(sarrera1)
        z2 = float(sarrera2)
    except ValueError:
        raise TypeError("Sartutako balioetako bat edo biak ez dira zenbakizkoak.")

    print(f"Zenbakien batura: {z1 + z2}")

try:
    batu_zenbakiak()
except TypeError as e:
    print(e)
```

## 5. ariketa

#### Idatzi Python programa bat, fitxategi bat ireki eta baimen arazoren bat badago PermissionError salbuespen bat kudeatzen duena.

```python

def irakurri_fitxategia_baimenekin():
    fitxategi_izena = input("Sartu ireki nahi duzun fitxategiaren bidea/izena: ")

    try:
        with open(fitxategi_izena, "r", encoding="utf-8") as fitxategia:
            edukia = fitxategia.read()
            print(f"\n--- '{fitxategi_izena}' fitxategiaren edukia ---")
            print(edukia)

    except PermissionError:
        print(f"\nErrorea: Ez duzu baimenik '{fitxategi_izena}' fitxategia irekitzeko edo irakurtzeko.")
    except FileNotFoundError:
        print(f"\nErrorea: Ezin izan da '{fitxategi_izena}' fitxategia aurkitu.")

irakurri_fitxategia_baimenekin()
```

## 6. ariketa

#### Idatzi Python programa bat zerrenda batean eragiketa bat egiten duena eta IndexError salbuespen bat kudeatzen duena indizea barrutitik kanpo badago.

```python
def lortu_elementua():
    fruituak = ["Sagarra", "Gozokia", "Platanoa", "Marrubia", "Mahatsa"]
    
    print("Zero zerrenda honako hau da:")
    print(fruituak)
    print(f"Index baliodunak: 0 eta {len(fruituak) - 1} artean (edo negatiboak -1etik -{len(fruituak)}-ra)\n")

    try:
        indizea = int(input("Sartu ikusi nahi duzun elementuaren indizea (zenbakia): "))
        
        elementua = fruituak[indizea]
        print(f"\nAukeratutako elementua (indizea {indizea}): {elementua}")

    except IndexError:
        print(f"\nErrorea: Indizea barrutitik kanpo dago! Zerrendak {len(fruituak)} elementu ditu bakarrik.")
        
    except ValueError:
        print("\nErrorea: Mesedez, sartu zenbaki oso baliodun bat.")

lortu_elementua()
```

## 7. ariketa

    #### Idatzi Python programa bat, erabiltzaileari zenbaki bat sartzeko eskatzen diona eta erabiltzaileak sarrera bertan behera uzten badu KeyboardInterrupt salbuespen bat kudeatzen duena.


```python
def eskatu_zenbakia_etenaldiarekin():
    try:
        print("Sartu zenbaki bat (edo sakatu 'Ctrl + C' programa bertan behera uzteko):")
        sarrera = input("> ")
        
        zenbakia = float(sarrera)
        print(f"\nEderki! Sartutako zenbakia: {zenbakia}")

    except KeyboardInterrupt:
        print("\n\n[Oharra] Prozesua bertan behera utzi duzu (KeyboardInterrupt). Agur!")

    except ValueError:
        print("\nErrorea: Sartutako testua ez da baliozko zenbakia.")

eskatu_zenbakia_etenaldiarekin()
```

## 8. ariketa

#### Idatzi zatiketa egiten duen eta errore aritmetiko bat badago ArithmeticError salbuespen bat kudeatzen duen Python programa bat.

```python

def zatiketa_aritmetikoa():
    try:
        zenbakigarria = float(input("Sartu zenbakigarria: "))
        zatitzailea = float(input("Sartu zatitzailea: "))
        
        emaitza = zenbakigarria / zatitzailea
        print(f"\nEmaitza: {emaitza}")

    except ArithmeticError as e:
        print(f"\n[Errore Aritmetikoa] Gertatutako errorea: {e}")
        
    except ValueError:
        print("\nErrorea: Mesedez, sartu baliozko zenbaki bat.")

zatiketa_aritmetikoa()
```

## 9. ariketa

#### Idatzi Python programa bat, fitxategi bat ireki eta kodeketa arazoren bat badago UnicodeDecodeError salbuespen bat kudeatzen duena.

```python

def irakurri_fitxategia_kodeketarekin():
    fitxategi_izena = input("Sartu irakurri nahi duzun fitxategiaren izena: ")

    try:
        with open(fitxategi_izena, "r", encoding="utf-8") as fitxategia:
            edukia = fitxategia.read()
            print(f"\n--- '{fitxategi_izena}' fitxategiaren edukia ---")
            print(edukia)

    except UnicodeDecodeError:
        print(f"\nErrorea: Ezin izan da '{fitxategi_izena}' fitxategia dekodetu. Ez dago UTF-8 formatuan kodedatuta.")
        print("Aholkua: Saiatu beste kodeketa batekin irekitzen (adibidez, encoding='latin-1').")

    except FileNotFoundError:
        print(f"\nErrorea: Ezin izan da '{fitxategi_izena}' fitxategia aurkitu.")

irakurri_fitxategia_kodeketarekin()
```

## 10. ariketa

#### Idatzi zerrenda-eragiketa bat egiten duen eta atributua existitzen ez bada AttributeError salbuespen bat kudeatzen duen Python programa bat.

```python

def atributu_errorea_kudeatu():
    zenbakiak = [10, 5, 20, 15, 30]
    
    print("Hasierako zerrenda:", zenbakiak)

    try:
        print("\n'sort_values()' metodoa deitzen saiatzen...")
        zenbakiak.sort_values()

    except AttributeError as e:
        print(f"\n[Harrapatutako Salbuespena]: AttributeError gertatu da.")
        print(f"Xehetasuna: {e}")
        
        print("\nOharra: Zerrendetarako 'sort()' metodoa erabili behar da.")
        zenbakiak.sort()
        print("Zuzendutako zerrenda ordenatua:", zenbakiak)

atributu_errorea_kudeatu()
```
