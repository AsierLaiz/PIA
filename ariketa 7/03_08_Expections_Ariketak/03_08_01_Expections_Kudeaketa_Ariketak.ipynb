# Exception Kudeaketa

Programatzen dugunean, arazoaren arabera, egoera berezi batzuk sor daitezke (sarrera-balio baliogabeak, zerrenda hutsak, etab.) exekuzioa eteten dutenak. Egoera horiek saihesteko ahalegina egin behar dugu, eta, ahal den neurrian, errore hauek exekuzio-garaian detektatu eta behar bezala kudeatu. Horretarako dira **Salbuespenak**.

Adibidez, funtzio honek bi zenbaki zatitzen ditu:


````python
def division(a, b):
    coc = a//b
    res = a % b
    return (coc, res)
````

````python
division(4,5)
````

Baina zer gertatzen da 0z zatitzea egiten saiatzen bagara?

````python
print(division(10, 0))
print(division(1024,10))
````

**Errorea!** Ezin duzu zenbaki bat 0z zatitu (emaitza infinitua da). Errore hau salbuespenak erabiliz kudeatu daiteke, behean erakusten den bezala:

````python
def division(a, b):
    try:
        coc = a//b
        res = a % b
        return (coc, res)
    except:
        print('Error en la división de {} entre {}.'.format(a, b))
````

Eta orain:

````python
print(division(10, 0))
print(division(1024,10))
````

Sintaxi osoagoa honako hau da:

```python
try:
    hemen doaz argibideak
except Exception_name:
    argibideen exekuzioa Exception_name salbuespena gertatu bada
except Exception_name2:
    argibideen exekuzioa Exception_name2 salbuespena gertatu bada
else:
    salbuespenik ez badago, kode bloke hau exekutatzen da
```

Beste egoera bat erabiltzaileak zenbakiak sartzen dituenean da. Zer gertatzen da datuak ez badira zenbakizkoak?

````python
def division(a, b):
    try:
        coc = a//b
        res = a % b
        return (coc, res)
    except:
        print('Error en la división de {} entre {}'.format(a, b))
````

````python

def main():
    num = int(input('Introduce un número:'))
    div = int(input('Introduce el divisor:'))
    print(division(num, div))

main() #posibles errores incluyen digitar texto en num o div, o definir div = 0
````

Kasu hauek kudeatzeko beste salbuespen bat gehitu behar dugu:

````python
def division(a, b):
    try:
        coc = a//b
        res = a % b
        return (coc, res)
    except ZeroDivisionError:
        print('La división por cero no está definida.')
        return ''


def main():
    try:
        num = int(input('Introduce un número:'))
        div = int(input('Introduce el divisor:'))
        print(division(num, div))
    except ValueError:
        print('El valor introducido no es un número.')
main()
````

## Finally

`finally` blokea zehazten da salbuespen bat gertatzen den ala ez kontuan hartu gabe exekutatu behar diren ekintzak zehazteko:

```python
try:
    # exekutatu ekintza hau lehenengo
except:
    # Exekutatu salbuespen bat gertatzen bada
finally:
    # Exekutatu beti kode hau
```

Salbuespenen exekuzio ordena honako hau da:

```try -> except -> else -> finally```

Adibidez:


````python
try:
    num = int(input("Introduce un número:"))
    re = 100/num
except:
    print("Ocurrió algún error :(")
else:
    print ("El resultado es: ",re)
finally :
    print ("Y por fin... el programa se acaba.")
````

Salbuespen mota zehaztea neketsua denez (batzuetan ez dakigu zer salbuespen mota gerta daitezkeen ere), kode lerro bakarra erabil daiteke (nahiz eta ahal den guztietan identifikatzea praktika ona den):

````python
try:
    num = int(input("Introduce el número:"))
    re = 100/num
    print (re)
except Exception as e :
    print (e, type(e))
````

Azkenik, salbuespenak jaurti daitezke ```raise``` erabiliz, erroreen kudeaketa kodean *gorago* kudea dadin:

````python
raise ZeroDivisionError('error de división por cero')
````

## Nire salbuespenak definitzea:

Gure salbuespenak definitzeko, Salbuespen klasetik heredatzen duen klase bat sortu behar dugu. Oraingoz, hurrengo adibidea txantiloi gisa ikus dezakezu.

````python
class NoPuedeIntroducirDosException(Exception):
    def __init__(self, value):
        self.value = value
        
    def __str__(self):
        return self.value
````

````python

def main():
    try:
        num = int(input())
        if num == 2:
            raise NoPuedeIntroducirDosException('¡El usuario introdujo 2 y no se puede!')
        else:
            print('Todo va bien, no se ha introducido el número 2.')
    except Exception as e:
        print(e, type(e))
        
main()
````

## 4. ariketa

Definitu salbuespen bat erabiltzaile batek 20 karaktere baino gehiago telefono zenbaki gisa sartzea eragozteko:

````python
agenda = {}

def process_numbers(person, number):
    agenda[person] = number

def main():
    n = int(input('Teclee el número de contactos:'))
    print('Agregando ' + str(n) + ' personas...')
    for i in range(n):
        nombre = input('Teclee el nombre del contacto {}:'.format(i+1))
        numero = int(input('Teclee el número de contacto {}:'.format(i+1)))
        process_numbers(nombre, numero)
    print('Así va la agenda:' + str(agenda))

main()
````
