## 1. Ariketa
### Idatzi 1 eta 10 arteko zenbaki oso bat eskatzen duen funtzio bat eta zenbaki horren biderketa taula gordetzen duena, non n sartutako zenbakia den, tabla-n.txt izeneko fitxategi batean


```python
def gorde_biderketa_taula():
    zenbaki = int(input("Sartu 1 eta 10 arteko zenbaki oso bat: "))
    
    if 1 <= zenbaki <= 10:
        # f-string erabiliz, 'zenbaki' aldagaiak fitxategiaren izena aldatzen du (adib. tabla-5.txt)
        fitxategi_izena = f"tabla-{zenbaki}.txt"
        
        with open(fitxategi_izena, "w") as fitxategia:
            for i in range(1, 11):
                emaitza = zenbaki * i
                fitxategia.write(f"{zenbaki} x {i} = {emaitza}\n")
        
        print(f"{zenbaki}(e)ren biderketa-taula '{fitxategi_izena}' fitxategian gorde da.")
    else:
        print("Errorea: Zenbakiak 1 eta 10 artekoa izan behar du.")

# Funtzioari dei egin
gorde_biderketa_taula()
```

## 2. Ariketa
### Idatzi 1 eta 10 arteko zenbaki oso bat eskatzen duen funtzio bat, zenbaki horren biderketa-taula duen tabla-n.txt fitxategia irakurtzen duena, non n sartutako zenbakia den, eta pantailan bistaratzen duena.
```python
def irakurri_biderketa_taula():
    zenbaki = int(input("Sartu 1 eta 10 arteko zenbaki oso bat: "))
    
    if 1 <= zenbaki <= 10:
        fitxategi_izena = f"tabla-{zenbaki}.txt"
        
        try:
            with open(fitxategi_izena, "r") as fitxategia:
                print(f"\n{zenbaki}(e)ren biderketa-taula ({fitxategi_izena}):")
                for lerroa in fitxategia:
                    print(lerroa.strip())
        except FileNotFoundError:
            print(f"Fitxategia ez da aurkitu: '{fitxategi_izena}' ez da existitzen.")
    else:
        print("Errorea: Zenbakiak 1 eta 10 artekoa izan behar du.")

# Funtzioari dei egin
irakurri_biderketa_taula()
```

## 3. Ariketa
### Idatzi 1 eta 10 artean n eta m bi zenbaki eskatzen dituen funtzio bat, zenbaki horren biderketa-taula duen tabla-n.txt fitxategia irakurtzen duena eta fitxategiaren m lerroa pantailan bistaratzen duena. Fitxategia existitzen ez bada, pantailan mezu bat erakutsi beharko luke erabiltzaileari horren berri emanez.

```python
def erakutsi_fitxategiaren_lerroa():
    # Bi zenbakiak eskatu erabiltzaileari
    n = int(input("Sartu 'n' zenbakia (1 eta 10 artean): "))
    m = int(input("Sartu 'm' lerro zenbakia (1 eta 10 artean): "))

    # Bi zenbakiak baliozko tartean daudela egiaztatu
    if (1 <= n <= 10) and (1 <= m <= 10):
        fitxategi_izena = f"tabla-{n}.txt"

        try:
            with open(fitxategi_izena, "r") as fitxategia:
                # Fitxategiko lerro guztiak zerrenda (list) batean kargatu
                lerroak = fitxategia.readlines()
                
                # 'm' lerroa erakutsi (kontuan izan zerrendak 0 indizetik hasten direla, beraz 'm - 1' erabili behar da)
                print(f"\n'{fitxategi_izena}' fitxategiko {m}. lerroa:")
                print(lerroak[m - 1].strip())

        except FileNotFoundError:
            print(f"Errorea: '{fitxategi_izena}' fitxategia ez da existitzen.")
    else:
        print("Errorea: Bi zenbakiek 1 eta 10 artekoak izan behar dute.")

# Funtzioari dei egin
erakutsi_fitxategiaren_lerroa()
```

```python
def salarie(hours, operator, rate):

    
    # Egiaztatu operadorea datu-basean dagoen
    if operator not in hours:
        print(f"Errorea: '{operator}' ez da aurkitu datu-basean.")
        return 0, 0.0

    # 1. Operadoreari dagozkion orduekin zerrenda bat sortu (balio hutsak baztertuz eta int-era bihurtuz)
    # .values() erabiltzen da hiztegiaren balioak lortzeko
    orduak = [int(i) for i in hours[operator].values() if i != '']

    # 2. Ordu guztiak batu (sum() funtzioa erabiliz)
    ordu_guztira = sum(orduak)

    # 3. Soldata kalkulatu ordu guztiak tarifarekin biderkatuz
    soldata = ordu_guztira * rate

    # Ordu guztiak eta soldata itzuli tupla batean
    return ordu_guztira, soldata


# --- ADIBIDEA (2. operadorearekin) ---

# Suposatuz 5.3 atalean honako hiztegia sortu duzula:
operadoreak = {
    'OP1': {'Urtarrila': '180', 'Otsaila': '160', 'Martxoa': '140', 'Apirila': '180'},
    'OP2': {'Urtarrila': '120', 'Otsaila': '140', 'Martxoa': '230', 'Apirila': '100'}
}

# Funtzioari dei egin (tarifari 10 balioa emanda)
orduak, soldata = salarie(operadoreak, 'OP2', 10)

# Emaitzak bistaratatu
print('Bigarren operadoreak lan egindako orduak:', orduak)
print('Bigarren operadorearen soldata:', soldata)
```

## 6. Ariketa

Idatzi programa bat enpresa baten bezeroen izen eta telefono zenbakiekin telefono-gida bat kudeatzeko. Programak funtzio hauek izan behar ditu: direktorio-fitxategia sortzeko, existitzen ez bada, bezero baten telefono-zenbakia bilatzeko, bezero berri baten telefono-zenbakia gehitzeko eta bezero baten telefono-zenbakia ezabatzeko. Direktorioa listin.txt testu-fitxategian gorde behar da, non bezeroaren izena eta telefono-zenbakia komaz bereizita agertu behar diren, bezero bakoitza lerro bereizi batean.

```python

import os

FITXATEGIA = "listin.txt"

# 1. Fitxategia sortu existitzen ez bada
def sortu_direktorioa():
    if not os.path.exists(FITXATEGIA):
        with open(FITXATEGIA, "w", encoding="utf-8") as f:
            pass  # Fitxategia hutsik sortzen du
        print(f"'{FITXATEGIA}' fitxategia sortu da.")
    else:
        print(f"'{FITXATEGIA}' jadanik existitzen da.")

# 2. Bezero baten telefonoa bilatu
def bilatu_telefonoa():
    izena = input("Sartu bilatu nahi duzun bezeroaren izena: ").strip()
    aurkitua = False
    
    try:
        with open(FITXATEGIA, "r", encoding="utf-8") as f:
            for lerroa in f:
                if lerroa.strip():
                    bezeroa, telefonoa = lerroa.strip().split(",")
                    if bezeroa.lower() == izena.lower():
                        print(f"-> {bezeroa}(ren) telefonoa: {telefonoa}")
                        aurkitua = True
                        break
        if not aurkitua:
            print(f"Ez da '{izena}' izeneko bezerorik aurkitu.")
    except FileNotFoundError:
        print(f"Errorea: '{FITXATEGIA}' ez da existitzen. Sortu lehenik.")

# 3. Bezero berria gehitu
def gehitu_bezeroa():
    izena = input("Sartu bezero berriaren izena: ").strip()
    telefonoa = input("Sartu bezeroaren telefono-zenbakia: ").strip()
    
    # Fitxategian bezeroa jadanik existitzen den egiaztatu
    existitzen_da = False
    try:
        with open(FITXATEGIA, "r", encoding="utf-8") as f:
            for lerroa in f:
                if lerroa.strip():
                    b_izena, _ = lerroa.strip().split(",")
                    if b_izena.lower() == izena.lower():
                        existitzen_da = True
                        break
    except FileNotFoundError:
        pass

    if existitzen_da:
        print(f"Errorea: '{izena}' jadanik erregistratuta dago.")
    else:
        # 'a' (append) moduan ireki amaieran idazteko
        with open(FITXATEGIA, "a", encoding="utf-8") as f:
            f.write(f"{izena},{telefonoa}\n")
        print(f"'{izena}' ondo gehitu da telefono-gidan.")

# 4. Bezero bat ezabatu
def ezabatu_bezeroa():
    izena = input("Sartu ezabatu nahi duzun bezeroaren izena: ").strip()
    ezabatua = False
    lerro_berriak = []
    
    try:
        with open(FITXATEGIA, "r", encoding="utf-8") as f:
            for lerroa in f:
                if lerroa.strip():
                    bezeroa, telefonoa = lerroa.strip().split(",")
                    if bezeroa.lower() == izena.lower():
                        ezabatua = True  # Lerro hau ez dugu berriz gordeko
                    else:
                        lerro_berriak.append(lerroa)
        
        if ezabatua:
            # Fitxategia berriro idatzi kentu den bezeroa gabe ('w' moduan)
            with open(FITXATEGIA, "w", encoding="utf-8") as f:
                f.writelines(lerro_berriak)
            print(f"'{izena}' ondo ezabatu da fitxategitik.")
        else:
            print(f"Ez da '{izena}' izeneko bezerorik aurkitu.")
            
    except FileNotFoundError:
        print(f"Errorea: '{FITXATEGIA}' ez da existitzen.")

# Erabiltzailearen menua
def menua():
    # Programaren hasieran fitxategia existitzen dela ziurtatu
    sortu_direktorioa()
    
    while True:
        print("\n--- TELEFONO-GIDA MENU-A ---")
        print("1. Bilatu bezero baten telefonoa")
        print("2. Gehitu bezero berria")
        print("3. Ezabatu bezero bat")
        print("4. Irten")
        
        Aukera = input("Aukeratu aukera bat (1-4): ").strip()
        
        if Aukera == "1":
            bilatu_telefonoa()
        elif Aukera == "2":
            gehitu_bezeroa()
        elif Aukera == "3":
            ezabatu_bezeroa()
        elif Aukera == "4":
            print("Eskerrik asko programa erabiltzeagatik!")
            break
        else:
            print("Aukera okerra. Sartu 1 eta 4 arteko zenbaki bat.")

# Programa abiarazi
menua()    
```
