### 1. Bi zerrenda hiztegi bihurtzea

Bi zerrenda emanda, idatzi Python programa bat hiztegi bihurtzeko, non list1-en elementua gakoa den eta list2-en elementua balioa.

keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]

Emaitza: {'Ten': 10, 'Twenty': 20, 'Thirty': 30}

```python
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]

hiztegia = dict(zip(keys, values))
print(hiztegia)
```

### 2. Bi Python hiztegi batu

dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = {'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

{'Ten': 10, 'Twenty': 20, 'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

```python
dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = {'Thirty': 30, 'Fourty': 40, 'Fifty': 50}

hiztegia_batua = {**dict1, **dict2}
print(hiztegia_batua)
```

### 4. Hasieratu hiztegia balio lehenetsiekin. Inprimatu 'Kelly' gakoaren balioa.

Erabili `fromkeys()` funtzioa

employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}

Emaitza: {'Kelly': {'designation': 'Developer', 'salary': 8000}, 'Emma': {'designation': 'Developer', 'salary': 8000}}


```python
employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}

hiztegia = dict.fromkeys(employees, defaults)
print(hiztegia)
```

### 5. Sortu hiztegi bat hiztegi jakin batetik gakoak ateraz

Idatzi Python programa bat hiztegi berri bat sortzeko, zehaztutako gakoak hiztegi honetatik ateraz.

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"}

keys = ["name", "salary"]

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"}

keys = ["name", "salary"]

hiztegia_berria = {gakoa: sample_dict[gakoa] for gakoa in keys}
print(hiztegia_berria)
```

### 6. Ezabatu hiztegi bateko gakoen zerrenda

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

#Keys to remove
keys = ["name", "salary"]

Erabili `pop()` metodoa

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

keys = ["name", "salary"]

for gakoa in keys:
    sample_dict.pop(gakoa)

print(sample_dict)
```

### 7. Hiztegi batean balio bat existitzen den egiaztatzea

sample_dict = {'a': 100, 'b': 200, 'c': 300}

```python
sample_dict = {'a': 100, 'b': 200, 'c': 300}

existitzen_da = 200 in sample_dict.values()
print(existitzen_da)
```

### 8. Hiztegi bateko gako-izena aldatzea

Idatzi programa bat hurrengo hiztegian "hiria" gakoa "kokapena" bihurtzeko.

sample_dict = {
  "name": "Kelly",
  "age":25,
  "salary": 8000,
  "city": "New york"
}

{'name': 'Kelly', 'age': 25, 'salary': 8000, 'location': 'New york'}

```python
sample_dict = {
  "name": "Kelly",
  "age": 25,
  "salary": 8000,
  "city": "New york"
}

sample_dict["location"] = sample_dict.pop("city")
print(sample_dict)
```

### 9. Lortu gutxieneko balioaren gakoa hiztegi honetatik

sample_dict = {
  'Physics': 82,
  'Math': 65,
  'history': 75
}

Erabili `min()` funtzioa

```python
sample_dict = {
  'Physics': 82,
  'Math': 65,
  'history': 75
}

gako_txikiena = min(sample_dict, key=sample_dict.get)
print(gako_txikiena)
```
