### 1. Tupla bat alderantzikatzea Pythonen

tuple1 = (10, 20, 30, 40, 50)

Emaitza: (50, 40, 30, 20, 10)

```python
tuple1 = (10, 20, 30, 40, 50)
tuple1 = tuple1[::-1]
print(tuple1)
```

### 5. Bi tupla trukatu Pythonen

### 6. Elementu espezifikoak tupla batetik tupla berri batera kopiatzea

### 7. Tupla aldatu
Tupla txertatu bat emanda, idatzi programa bat tupla bateko zerrenda baten lehen elementua (22) 222ra aldatzeko.

tuple1 = (11, [22, 33], 44, 55)

Emaitza: tuple1: (11, [222, 33], 44, 55)

```python
tuple1 = (11, [22, 33], 44, 55)
tuple1[1][0] = 222
print("tuple1:", tuple1)
```

### 8. Ordenatu tupla bat bigarren elementuaren arabera

tuple1 = (('a', 23),('b', 37),('c', 11), ('d',29))

Emaitza: 

(('c', 11), ('a', 23), ('d', 29), ('b', 37))

```python
tuple1 = (('a', 23), ('b', 37), ('c', 11), ('d', 29))
tuple1 = tuple(sorted(tuple1, key=lambda elementua: elementua[1]))
print(tuple1)
```

### 9. Tupla baten 50. elementuaren agerpen kopurua zenbatu

tuple1 = (50, 10, 60, 70, 50)

Erabili `count()` funtzioa

```python
tuple1 = (50, 10, 60, 70, 50)
kopurua = tuple1.count(50)
print(kopurua)
```

### 10. Egiaztatu tuplaren elementu guztiak berdinak diren ala ez

tuple1 = (45, 45, 45, 45)

```python
tuple1 = (45, 45, 45, 45)
guztiak_berdinak = all(elementua == tuple1[0] for elementua in tuple1)
print(guztiak_berdinak)
```
