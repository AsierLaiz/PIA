### 2. Bi zerrenda indizearen arabera kateatu
Idatzi programa bat bi zerrenda indizearen arabera batzeko. Sortu zerrenda berri bat jatorrizko zerrendaren 0 indizean dagoen elementua duena, gero 1 indizean dagoen elementua, eta horrela azken elementura arte. Geratzen diren elementuak zerrenda berriaren amaieran gehituko dira.

Erabili `zip()` funtzioa

list1 = ["M", "na", "i", "Ke"]
list2 = ["y", "me", "s", "lly"]

Emaitza: 
['My', 'name', 'is', 'Kelly']

### 3. Zerrenda bateko elementu bakoitza bere karratu bihurtu
Zenbaki zerrenda bat emanda, idatzi programa bat zerrendako elementu bakoitza bere karratu bihurtzeko.

numbers = [1, 2, 3, 4, 5, 6, 7]

Emaitza: [1, 4, 9, 16, 25, 36, 49]

### 4. Kateatu bi zerrenda hurrengo ordenan

### 5. Bi zerrendak aldi berean iteratu

list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

Emaitza: 

10 400

20 300

30 200

40 100

### 7. Gehitu elementu berri bat zerrendara elementu espezifiko baten ondoren

Idatzi Python programa bat hurrengo zerrendako 6000. elementuaren ondoren 7000. elementua gehitzeko

list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]

Emaitza: [10, 20, [300, 400, [5000, 6000, 7000], 500], 30, 40]

### 8. Zerrenda txertatu bat luzatzea azpizerrenda bat gehituz

Zerrenda txertatu bat emanda, idatzi programa bat hura luzatzeko ["h", "i", "j"] azpizerrenda gehituz, hurrengo zerrendaren itxura izan dezan.

list1 = ["a", "b", ["c", ["d", "e", ["f", "g"], "k"], "l"], "m", "n"]

Gehitzeko azpizerrenda: sub_list = ["h", "i", "j"]

Emaitza: ['a', 'b', ['c', ['d', 'e', ['f', 'g', 'h', 'i', 'j'], 'k'], 'l'], 'm', 'n']

### 9. Ordeztu zerrenda-elementua balio berri batekin, aurkitzen bada

Python zerrenda bat emanda, idatzi programa bat zerrendan 20 balioa aurkitzeko eta, baldin badago, 200-rekin ordezkatzeko. Eguneratu elementuaren lehen agerpena bakarrik.

list1 = [5, 10, 15, 20, 25, 50, 20]

Emaitza: [5, 10, 15, 200, 25, 50, 20]
