# 2. ariketa — Modulua inportatzea VS Kodea Exekutazea

Aldatu aurreko ariketako converter.py modulua fitxategiaren amaieran proba-lerro bat gehituz: `print("Converter modulua kargatuta").`

Egiaztatu zer gertatzen den bi aldiz inportatzen duzunean (adibidez, bi gelaxka desberdinetatik) eta azaldu iruzkin batean zergatik `print` adierazpen hau lehenengo aldian bakarrik exekutatzen den.

Ondoren, zuzendu modulua `if __name__ == "__main__":` eraikuntza erabiliz, mezu hau fitxategia zuzenean exekutatzen bada bakarrik bistara dadin, eta ez inportatzen denean.

```python
import converter  # lehen inportazioa -> mezua behin bistaratzen da
import converter  # bigarren inportazioa (beste gelaxka batetik ere berdin gertatuko litzateke) -> mezua EZ da berriro agertzen

import sys
print("converter" in sys.modules)
```

## Azalpena: 

Python-ek modulu bat inportatzen den lehen aldian bakarrik exekutatzen du haren goi-mailako kodea (funtzio eta klase definizioetatik kanpo dagoena, kasu honetan `print("Converter modulua kargatuta")`). Modulua behin kargatu ondoren, Python-ek modulu-objektua `sys.modules` hiztegian gordetzen du cache moduan. Ondorengo `import converter` guztiek (gelaxka berean zein beste gelaxka batzuetatik) ez dute fitxategia berriz exekutatzen; `sys.modules`-en dagoen objektu bera itzultzen dute besterik gabe. Horregatik agertzen da `print` mezua behin bakarrik, inportazio kopurua edozein dela ere.

```python
import importlib
import converter

importlib.reload(converter)  

import subprocess

emaitza = subprocess.run(["python3", "converter.py"], capture_output=True, text=True)
print(emaitza.stdout) 
```

# 3. ariketa — Erroreen kudeaketa duen modulua

```python
import bank

print(bank.erretiratu(1000, 200))    
print(bank.erretiratu(1000, 1500))   
print(bank.erretiratu(1000, -50))    
```
