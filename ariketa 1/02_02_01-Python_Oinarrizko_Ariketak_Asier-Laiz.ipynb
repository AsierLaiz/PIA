## 3.2 while

### Ariketa 27

```python
zenbaki=input("Sartu zenbaki bat")
while zenbaki!="0":
    print("Sartutako zenbakia 0 ez da")
    zenbaki=input("Sartu zenbaki bat")
```

### Ariketa 28

Eskatu erabiltzaileari zenbakiak sartzeko, guztien baturak 100 zenbakia gainditu arte:

```python
zenbaki=input("Sartu zenbaki bat")
suma=0
while suma < 100:
    suma += int(zenbaki)
    print("Suma gaurko zenbakiarekin: ", suma)
    if suma >= 100:
        print("Suma 100 baino handiagoa da, programa amaitzen da.")
        break
    zenbaki=input("Sartu zenbaki bat")
```

### Ariketa 29

Eskatu erabiltzaileari zenbakiak sartzeko, 0 bat sartu arte. Sartutako zenbakien baturak begiztaren 100 gatz balioa bortxaz gainditzen badu:

```python
batura = 0

while True:
    try:
        zenbakia = float(input("Sartu zenbaki bat (0 amaitzeko): "))

        if zenbakia == 0:
            print("0 sartu duzu. Programa amaitu da.")
            break

        batura += zenbakia
        print(f"Uneko batura: {batura}")

        if batura > 100:
            print(
                "\nBaturak 100 balioa gainditu du! Begizta bortxaz gelditzen da."
            )
            break

    except ValueError:
        print("Errorea: Sartu baliodun zenbaki bat.")

print(f"\nAzken batura: {batura}")
```

### Ariketa 30

Eskatu erabiltzaileari gero eta zenbaki handiagoak sartzeko iterazio bakoitzean. Aurrekoaren berdina edo txikiagoa den zenbaki bat sartzen baduzu, buka ezazu begizta. Gainera, erabiltzaileak beti izango du begiztaren amaiera 0 balioaren bidez behartzeko aukera. Begizta behar bezala amaitu bada (behartu gabe), "END" mezua erakutsiko du:

```python
azken_zenbakia = None

while True:
    zenbakia = float(input("Sartu zenbaki bat (0 amaitzeko): "))

    if zenbakia == 0:
        print("Amaituta.")
        break

    if azken_zenbakia is not None and zenbakia <= azken_zenbakia:
        print("Ez da aurrekoa baino handiagoa.")
        print("END")
        break

    azken_zenbakia = zenbakia
```

### Ariketa 35

Begiratu "cars" hiztegiko elementu guztiak, eta erakutsi gakoa "S" letrarekin hasten den edo autoa "M" letrarekin hasten den. Begizta bortxaz amaitzen da "C" letrarekin hasten den autoren bat aurkitzen badu:

```python
autoak = {"Seat": "Leon", "Mercedes": "Clase A", "Citroen": "C3", "Mazda": "MX-5", "Renault": "Megane"}

for gakoa, autoa in autoak.items():
    if gakoa.startswith("S"):
        print(gakoa)
    if autoa.startswith("M"):
        print(autoa)
    if autoa.startswith("C"):
        print("C letrarekin hasten den autoa aurkitu da, begizta bortxaz amaitzen da.")
        break
```

# 4. Funtzioak

### Ariketa 36

Sarrerako testu bat onartzen duen eta barruan duen bokal kopurua itzultzen duen funtzio bat definitzen du:

```python
def bokal_kopurua(testua):
    bokalak = "aeiouAEIOU"
    kopurua = 0
    for karakterea in testua:
        if karakterea in bokalak:
            kopurua += 1
    return kopurua

testua = input("Sartu testu bat: ")
print(bokal_kopurua(testua))
```

### Ariketa 37

Zerrenda bat parametro gisa onartzen duen eta pantailako zerrendan dauden testuetako bokal kopurua erakusten duen funtzio bat definitzen du:

```python
def bokal_kopurua(testua):
    bokalak = "aeiouAEIOU"
    kopurua = 0
    for karakterea in testua:
        if karakterea in bokalak:
            kopurua += 1
    return kopurua

def zerrendako_bokalak(zerrenda):
    for testua in zerrenda:
        print(testua, "-", bokal_kopurua(testua), "bokal")

testuak = ["Kaixo", "Python", "Programazioa"]
zerrendako_bokalak(testuak)
```
