## 2. Datu-bildumak

### 2.1 Zerrendak

Zerrendak dira datu-bildumarik erabilienak.

```python
mi_lista = ["primero", "segundo", "tercero"]
print(mi_lista)
```

Zure datuak **indexatuta eta ordenatuta** daude, datu bakoitzak posizio zehatz bat duela ulertuta. Halaber, definitutako zerrenda baten posizio zehatzak atzitu eta alda ditzakegu:

```python
# Zerrenda abiaraziko dugu.
mi_lista = ["primero", "segundo", "tercero"]
# Aldatu 2. indizean gordetako balioa (gogoratu 0 indizetik hasten direla zenbatzen).
mi_lista[2] = "cuarto"
print(mi_lista)
```

Elementu berriak ere gehitu daitezke, edo horietakoren bat ezabatu, baita beste elementu batzuen zerrenda egin ere:

```python
mi_lista = ["primero", "segundo", "tercero"]
print(mi_lista)

#Añadir un nuevo elemento en una posición concreta, en este caso al principio y al final
mi_lista.insert(0, "nuevo_elemento_inicio")
mi_lista.insert(len(mi_lista), "nuevo_elemento_final")
print(mi_lista)

#Eliminar un elemento por su valor
mi_lista.remove("nuevo_elemento_inicio")
print(mi_lista)

#Eliminar un elemento por su posición
mi_lista.pop(3)
print(mi_lista)

#Extender una lista con los valores de otra
mi_segunda_lista = ["cuarto", "quinto", "sexto"]
mi_lista.extend(mi_segunda_lista)
print(mi_lista)
```

### 2.4 Hiztegiak

Hiztegiak informazioa klabe-balio bikoteetan gordetzeko erabiltzen dira klabea:balioa

```python
mi_diccionario = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(mi_diccionario)
```

Hiztegietan ez da onartzen *gakoa* elementuan bikoiztuta, baina bai *balioa* elementuan. Gainera, **edozein motatako balioak eta gakoak erabil daitezke**:

```python
mi_diccionario = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
    1: "Mustang"
}
print(mi_diccionario)
```

Hiztegiak **alda daitezke**, edozein gakori lotutako balioa alda daiteke, baina ez gakoak berak:

```python
mi_diccionario = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mi_diccionario["year"] = 2018
print(mi_diccionario)
```

Elementu berriak gehi daitezke edo dagoen elementuren bat ezaba daiteke, honela:

```python
# Añadir nuevos elementos:
mi_diccionario = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mi_diccionario["color"] = "red"
print(mi_diccionario)

# Eliminar un elemento concreto por la clave:
mi_diccionario.pop("color")
print(mi_diccionario)
```

Hiztegi bateko elementuak, giltzarriak zein balioak, zeharkatzerakoan eta erabiltzerakoan, hainbat metodo erabil ditzakegu:

```python
mi_diccionario = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# FOR: Kasu honetan, "x" aldagaiak hiztegiko gakoak gordeko ditu begiztaren iterazio bakoitzean
for x in mi_diccionario:
  # En este caso, se utiliza la convesión a string sobre los valores del diccionario para evitar errores a la hora de concatenar (no se puede concatenar un string con un valor entero)
  print("[" + x + "] = " + str(mi_diccionario[x]))

# ".keys ()" eta ".values ()" funtzioak ere erabil ditzakegu gako eta balioen zerrenda bat lortzeko, hurrenez hurren.
claves = mi_diccionario.keys()
valores = mi_diccionario.values()
print(claves)
print(valores)

#Biak aldi berean ibili nahi izanez gero, honela egingo litzateke:
for x, y in mi_diccionario.items():
  # Siendo la "x" la clave y la "y" el valor de cada elemento del diccionario
  print(x, y)

```

## 3. Kontrol-egiturak

### 3.1. if..else

Python-en, hauek dira konparazio-adierazpen logiko nagusiak:

* Berdin: **a = = b**
* Desberdina: **a! = b**
Adingabea: **a < b**
* Txikiagoa edo berdina: **a < = b**
* Nagusia: **a > b**
* Handiagoa edo berdina: **a > = b**

Gainera, python beste lengoaia batzuetako "{}" giltza tipikoen ordez ***sangria** erabiltzean oinarritzen da, kontrol-egitura baten barruan dauden jarraibideak bereizteko:

```python
a = 41
b = 12
if a > b:
  print("a es mayor que b")
else:
  print("b es mayor que a")
```

Beste lengoaia batzuetan bezala, *if* instrukzioa soilik erabil daiteke habia egin gabe, edo hainbat habia egin daitezke **elif** etiketaren bidez:

```python
a = 41
if a > 10:
  print("a es mayor que 10")
  if a > 50:
    print("a es mayor que 50")
  elif a > 40:
    print("a es mayor que 40")
  else:
    print("a es menor de 40")
print("FIN")
```

*if* baldintza baten barruan kode-lerro bakarra exekutatuko badugu, lerro bakar batean idazteko aukera izango dugu, gure kodea murrizteko:

```python
a = 21
b = 10
if a > b: print("a es mayor que b")

# if..else adibide bat
print("AAA") if a > b else print("BBB")

#  if..elif..else adibidea
print("A > B") if a > b else print("A = B") if a == b else print("B > A")
```

Beste aukera bat da hainbat elementu aldi berean alderatzea **and** eta **or** eragigaiak erabiliz, konparazioak kateatzeko:

```python
a = 21
b = 10
c = 17

if a > b or a > c: print("a no es el elemento más pequeño")

if a > b and a > c: print("a es elemento con el mayor valor")
elif b > a and b > c: print("b es el elemento con el mayor valor")
elif c > a and c > b: print("c es el elemento con el mayor valor")
else: print("hay al menos dos elementos con los valores más altos")
```

### 3.2. while

*while* egiturak, beste lengoaia batzuetan bezala, haren ***edukia exekutatzen du, baldin eta baldintza batek*** egiazkoa izaten jarraitzen badu:

```python
i = 1
while i < 6:
  print(i)
  i += 1
```

Nola ez,**break** aginduaz ere balia gaitezke begizta amaitzeko interesatzen zaigunean, eta **continue**, berriz, hurrengo iteraziora pasatzeko:

```python
i = 0
while i < 10:
  i += 1
  if i == 3: continue
  print(i)
  if i == 5: break
```

Azkenik, Python-en, **else** sententzia *begizta while* baten amaieran erabiltzeko aukera sartu da, kode-bloke bat exekutatzeko, baldintzak egiazkoa izateari uzten dionean:

```python
i = 0
while i < 5:
  i+=1
  print(i)
else:
  print("Hemos terminado el bucle III")

# Este otro else no se ejecutará ya que no ha dejado de ser verdadera la condición:
j = 0
while j < 5:
  j += 1
  print(j)
  break
else:
  print("Hemos terminado el bucle JJJ")
```

### 3.3. for

Azken kontrol-egitura **for** da. Egitura hori hurrengo egoeratan erabil daiteke:
* Zerrenda bat iteratu
* String bat iteratu

```python
# Zerrenda bat iteratu
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)

print("---------------------------")
# String bat iteratu
for x in "banana":
  print(x)
```


*while* rekin bezala, **break** jarraibideak erabili ahal izango ditugu begizta amaitzeko, eta **continue**, hurrengo interaziora pasatzeko:

```python
for x in "botella":
  if x == "o": continue
  print(x)
  if x == "l": break

print("FIN")
```

Halaber, lehen bezala, **else** instrukzioa erabili ahal izango da azken bloke bat exekutatzeko, *for* instrukzioa behar bezala exekutatzen amaitu denean (*break* instrukzioa erabili gabe):

```python
for x in range(3):
  print(x)
else:
  print("for terminado correctamente")


for x in range(10, 16, 2):
  print(x)
  if x == 12: break
else:
  print("ESTE TEXTO NO SE MOSTRARÁ")
```

## 4. Funtzioak

Funtzioak definitzeko, **def** erabiliko dugu:

```python
def mi_funcion():
  print("Código interno de la función")

# Funtzioa deitu
mi_funcion()
```

Funtzioari parametroak pasatu nahi badizkiogu, honela egingo dugu:

```python
def mi_funcion(param):
  print("El parámetro recibido es: " + str(param))

mi_funcion("test")
mi_funcion(2)
```

Hasiera batean, funtzio bati definituta dituen parametro kopuru berarekin deitu behar zaio. Beraz, bi parametro baditu, bi parametrorekin deituko diogu, ez bat ez hiru. Hala ere, funtzio bat defini dezakegu parametro-kopurua jakin gabe, kontuan izanik zerrenda batean jasoko dituela:

```python
def mi_funcion(*params):
  print(type(params))
  print("He recibido " + str(len(params)) + " parámetros")
  if len(params) > 0: print("El primer parámetro es: " + str(params[0]))
  
  # Aquí recorro y muestro todos los parámetros
  for x in params:
    print(x)

mi_funcion()
print("-----------------")
mi_funcion("Primero", 23, "Tercero")
```

Era berean, argumentuak nahiago dugun ordenan aurkez ditzakegu:

```python
def mi_funcion(x, y, z):
  print("x = " + str(x))
  print("y = " + str(y))
  print("z = " + str(z))

mi_funcion(z = "a", x = 2, y = "YYY")
```

Aurreko bi kontzeptuak konbinatzen baditugu, edozein parametro-kopuru lor dezakegu, edozein ordenatan, funtziorako hiztegi bat izan dadin:

```python
def mi_funcion(**params):
  print("El apellido es: " + params["apellido"])
  print(params)

mi_funcion(apellido = "Jorge", nombre = "Mikel", email = "mjorgesote@educacion.navarra.es")

#Iristen den ala ez ez dakigun parametro bat eskuratzeko arriskua benetan ez jasotzea eta errore bat izatea da.
#Aipatu beheko deia egiaztatzeko:
#mi_funcion (nombre = "Mikel", email = "mjorgesote@educacion.navarra.es")
```

Parametro bakoitzerako lehenetsitako balio bat ere definitu ahal izango dugu, eta horrela ez badator, balio hori lehenetsita hartuko da:

```python
def mi_funcion(x = 10, y = "YYY"):
  print(x, y)

mi_funcion(222)
print("-----------------")
mi_funcion(y = "ZZZ")
print("-----------------")
mi_funcion()
```

Amaitzeko, beti dago funtzio baten emaitza lortu nahi izatea. Horretarako, **return** instrukzioa erabil dezakegu edozein elementu itzultzeko

```python
def mi_funcion(x):
  return x * 5

print(mi_funcion(4))
```
