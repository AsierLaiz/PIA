# Modularidad: módulos y paquetes

**Cómo trabajar desde el notebook:**

- Para crear un fichero `.py`, escribe en la **primera línea de la celda** `%%writefile nombre.py` y debajo el contenido del módulo. (Con paquetes: `%%writefile carpeta/nombre.py`.)
- Para crear una carpeta (paquete): `import os` y `os.makedirs("carpeta", exist_ok=True)`.
- Si cambias un módulo que ya importaste: **Kernel > Restart** (o `importlib.reload(modulo)`), si no seguirá usando la versión antigua.
- Para ejecutar un fichero como programa: `%run main.py`.
