## Ejercicio 1: Notas de alumnos

Tienes este diccionario con las notas de cada alumno:

```python
notas = {"Ane": [7, 8, 6], "Jon": [4, 3, 5], "Leire": [], "Mikel": [9, 10, 8]}
```

a) Crea una función `medias(d)` que reciba el diccionario y devuelva otro diccionario `{alumno: media}`. Si un alumno no tiene notas, su media será `None` (usa `try/except` con `ZeroDivisionError`).
b) Muestra el alumno con la mejor media y su media.
c) Muestra una lista con los alumnos suspendidos (media menor que 5).
d) Muestra una lista con los alumnos que no tienen notas.

````python
notas = {"Ane": [7, 8, 6], "Jon": [4, 3, 5], "Leire": [], "Mikel": [9, 10, 8]}
def medias(d):
    resultado = {}
    for alumno, lista in d.items():
        try:
            resultado[alumno] = sum(lista) / len(lista)
        except ZeroDivisionError:
            resultado[alumno] = None
    return resultado

m = medias(notas)
print(m)
validas = {a: v for a, v in m.items() if v is not None}

print(max(validas, key=validas.get))

suspenidos=[a for a, v in validas.items() if v<5 ]
print(suspenidos)


sin_notas = [a for a, v in m.items() if v is None]
print("Sin notas:", sin_notas)
````

## Ejercicio 2: Inventario de una tienda

Tienes una lista de tuplas `(producto, precio, stock)`:

```python
productos = [("pan", 1.2, 10), ("leche", 0.9, 0), ("queso", 6.5, 3), ("huevos", 2.4, 0)]
```

a) Crea una función `valor_total(lista)` que devuelva el valor total del inventario (suma de precio x stock de cada producto).
b) Muestra una lista con los nombres de los productos sin stock.
c) Convierte la lista en un diccionario `{producto: (precio, stock)}`.
d) Pide al usuario el nombre de un producto y muestra su precio y su stock. Si no existe, captura el `KeyError` y avisa.

````python
productos = [("pan", 1.2, 10), ("leche", 0.9, 0), ("queso", 6.5, 3), ("huevos", 2.4, 0)]

def valor_total(lista):
    total = 0
    for p in lista:
        total += p[1]*p[2]
    return total
print("Valor total:", valor_total(productos))

sin_stock = [p[0] for p in productos if p[2] == 0]
print("Sin stock:", sin_stock)

catalogo = {nombre:(precio, stock) for nombre,precio, stock in productos}

nombre = input("Producto: ")
try:
    precio, stock = catalogo[nombre]
    print("Precio:", precio, "- Stock:", stock)
except KeyError:
    print("Ese producto no existe")
````

## Ejercicio 3: Agenda de contactos con menú

Crea un programa con una agenda (diccionario `{nombre: telefono}`) que muestre este menú en un bucle hasta que el usuario elija salir:

```
1. Añadir contacto
2. Buscar contacto
3. Borrar contacto
4. Salir
```

- Opción 1: pide nombre y teléfono y lo guarda.
- Opción 2: pide un nombre y muestra su teléfono. Si no existe, captura el `KeyError` y avisa.
- Opción 3: pide un nombre y lo borra. Si no existe, captura el `KeyError` y avisa.
- Cualquier otra opción: muestra "Opción no válida".

````python
agenda = {}

while True:
    print("1. Añadir contacto")
    print("2. Buscar contacto")
    print("3. Borrar contacto")
    print("4. Salir")
    opcion = input("Opción: ")
    if opcion=="1":
        nombre=input("nombre")
        tel=input("telefono")
        agenda[nombre] = tel
    elif opcion=="2":
        nombre=input("nombre")
        try:
            print(agenda[nombre])
        except KeyError:
            print("pos no")
    elif opcion=="3":
        nombre=input("nombre")
        try:
            del agenda[nombre]
        except KeyError:
            print("pos tampoco")
    else :
        print("opcion no valida")

````

## Ejercicio 4: Alumnos desde un fichero

Crea un fichero `alumnos.csv` en la misma carpeta que el notebook con este contenido (formato `nombre;edad;nota`):

```
Ane;19;7.5
Jon;17;4
Leire;20;abc
Mikel;22;8
```

a) Lee el fichero y guarda cada línea como una tupla `(nombre, edad, nota)` en una lista (edad como `int`, nota como `float`). Si una línea tiene un dato que no se puede convertir, captura el `ValueError`, muestra un aviso con esa línea y pasa a la siguiente.
b) Si el fichero no existe, captura el `FileNotFoundError` y muestra un mensaje.
c) Crea una lista con los alumnos mayores de edad (18 o más) que han aprobado (nota 5 o más).
d) Escribe en un fichero `aprobados.txt` el nombre de cada alumno de esa lista, uno por línea.

````python
alumnos = []

try:
    with open("alumnos.csv", "r", encoding="utf-8") as f:
        for linea in f:
            campos = linea.strip().split(";")
            try:
                alumnos.append((campos[0], int(campos[1]), float(campos[2])))
            except ValueError:
                print("Línea con error:", linea.strip())
except FileNotFoundError:
    print("El fichero alumnos.csv no existe")

lista =[a for a in alumnos if a[1]>=18 and a[2]>=5]
print (lista)
with open("aprobados.txt", "w", encoding="utf-8") as f:
    for a in lista:
        f.write(a[0] + "\n")
````
