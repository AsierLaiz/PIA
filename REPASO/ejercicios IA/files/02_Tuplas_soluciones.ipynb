# Tuplas - Soluciones

## Ejercicio 1

Dada `t = (5, 8, 5, 2, 9, 5)`, muestra la tupla invertida y cuántas veces aparece el primer elemento.

```python
t = (5, 8, 5, 2, 9, 5)

print(t[::-1])
print(t.count(t[0]))
```

## Ejercicio 2

Dada `t = (("Ane", 7), ("Jon", 4), ("Mikel", 9))`, ordénala por la nota (segundo elemento) de mayor a menor.

```python
t = (("Ane", 7), ("Jon", 4), ("Mikel", 9))

ordenada = tuple(sorted(t, key=lambda e: e[1], reverse=True))
print(ordenada)
```

## Ejercicio 3

Crea una función que reciba dos números y devuelva una tupla con su suma y su producto. Recoge el resultado en dos variables y muéstralas.

```python
def suma_producto(a, b):
    return a + b, a * b

suma, producto = suma_producto(3, 4)
print(suma)
print(producto)
```

## Ejercicio 4

Dada `t = (1, [2, 3], 4)`, cambia el 3 por un 30. Después, elimina el 4 de la tupla (recuerda que son inmutables).

```python
t = (1, [2, 3], 4)

# El interior es una lista, así que sí se puede cambiar
t[1][1] = 30
print(t)

# Para quitar el 4: tupla -> lista -> quitar -> tupla
l = list(t)
l.remove(4)
t = tuple(l)
print(t)
```
