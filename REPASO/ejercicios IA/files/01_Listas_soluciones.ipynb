# Listas - Soluciones

## Ejercicio 1

Dada `numeros = [4, 9, 2, 7, 9, 1]`, crea una lista nueva con los números mayores que 3 y muéstrala invertida.

```python
numeros = [4, 9, 2, 7, 9, 1]

mayores = [n for n in numeros if n > 3]
print(mayores[::-1])
```

## Ejercicio 2

Pide al usuario 5 nombres, guárdalos en una lista y muestra cuántos empiezan por vocal.

```python
nombres = []
for i in range(5):
    nombres.append(input(f"Nombre {i + 1}: "))

vocales = "aeiouAEIOU"
contador = len([n for n in nombres if n and n[0] in vocales])
print("Empiezan por vocal:", contador)
```

## Ejercicio 3

Dada `palabras = ["kaixo", "mundua", "ni", "izena"]`, crea otra lista con la longitud de cada palabra (usa una comprensión).

```python
palabras = ["kaixo", "mundua", "ni", "izena"]

longitudes = [len(p) for p in palabras]
print(longitudes)
```

## Ejercicio 4

Dada `l = [10, 20, 30, 20, 40]`, elimina todas las apariciones de 20 y añade al final el valor que introduzca el usuario. Muestra la lista final.

```python
l = [10, 20, 30, 20, 40]

l = [e for e in l if e != 20]
l.append(input("Valor a añadir: "))
print(l)
```

## Ejercicio 5

Dadas `l1 = [1, 2, 3, 4]` y `l2 = [10, 20, 30, 40]`, crea una lista nueva con la suma de los elementos que ocupan la misma posición.

```python
l1 = [1, 2, 3, 4]
l2 = [10, 20, 30, 40]

sumas = [a + b for a, b in zip(l1, l2)]
print(sumas)
```
