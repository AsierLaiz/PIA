## Ejercicio 1

Crea `es_par(n)` que devuelva `True` o `False`. Úsala para filtrar los números pares de `[1, 2, 3, 4, 5, 6]`.

```python
def es_par(n):
    return n % 2 == 0

numeros = [1, 2, 3, 4, 5, 6]
pares = [n for n in numeros if es_par(n)]
print(pares)
```

## Ejercicio 2

Crea `media(*numeros)` que acepte cualquier cantidad de argumentos y devuelva la media. Pruébala con 2 y con 5 números.

```python
def media(*numeros):
    return sum(numeros) / len(numeros)

print(media(4, 6))
print(media(1, 2, 3, 4, 5))
```

## Ejercicio 3

Crea `saludar(nombre, saludo="Kaixo")` que muestre el saludo con el nombre. Llámala con y sin el segundo argumento.

```python
def saludar(nombre, saludo="Kaixo"):
    print(saludo, nombre)

saludar("Asier")
saludar("Asier", "Hola")
```

## Ejercicio 4

Crea una función recursiva que calcule el factorial de n. Pruébala con 5.

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## Ejercicio 5

Crea una función que reciba una lista y devuelva una tupla con el mayor y el menor elemento. Recoge el resultado en dos variables.

```python
def mayor_menor(lista):
    return max(lista), min(lista)

mayor, menor = mayor_menor([4, 6, 8, 24, 12, 2])
print(mayor)
print(menor)
```
