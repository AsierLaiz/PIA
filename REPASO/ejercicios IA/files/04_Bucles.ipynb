# Bucles (for / while)

Escribe tu solución en el bloque de código de cada ejercicio.

## Ejercicio 1

Pide números al usuario hasta que escriba 0. Al final, muestra la suma y cuántos números ha introducido.

## Ejercicio 2

Muestra con un `for` los múltiplos de 7 entre 1 y 100.

## Ejercicio 3

Pide una contraseña hasta que sea correcta o hasta que falle 3 veces (usa `while` y `break`). Muestra si ha acertado o si se ha quedado sin intentos.

## Ejercicio 4

Dada `l = [3, 8, 1, 9, 4]`, recórrela con `for i in range(len(l))` y muestra la posición y el valor de cada elemento.
