## Ejercicio 1

Pide números al usuario hasta que escriba 0. Al final, muestra la suma y cuántos números ha introducido.

```python
suma = 0
x = 1
while x != 0:
    x = int(input("sartu zenbaki: "))
    suma += x

print(suma)

```

## Ejercicio 2

Muestra con un `for` los múltiplos de 7 entre 1 y 100.

```python
zazpimult=[]
for i in range(1,100):
    if i % 7 == 0:
        zazpimult.append(i)
print(zazpimult)

```

## Ejercicio 4

Dada `l = [3, 8, 1, 9, 4]`, recórrela con `for i in range(len(l))` y muestra la posición y el valor de cada elemento.
