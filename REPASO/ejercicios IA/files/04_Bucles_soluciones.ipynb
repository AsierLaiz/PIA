# Bucles (for / while) - Soluciones

## Ejercicio 1

Pide números al usuario hasta que escriba 0. Al final, muestra la suma y cuántos números ha introducido.

```python
suma = 0
cantidad = 0

while True:
    n = float(input("Número (0 para terminar): "))
    if n == 0:
        break
    suma += n
    cantidad += 1

print("Suma:", suma)
print("Números introducidos:", cantidad)
```

## Ejercicio 2

Muestra con un `for` los múltiplos de 7 entre 1 y 100.

```python
for i in range(1, 101):
    if i % 7 == 0:
        print(i)
```

## Ejercicio 3

Pide una contraseña hasta que sea correcta o hasta que falle 3 veces (usa `while` y `break`). Muestra si ha acertado o si se ha quedado sin intentos.

```python
correcta = "1234"
intentos = 0
acertada = False

while intentos < 3:
    intento = input("Contraseña: ")
    if intento == correcta:
        acertada = True
        break
    intentos += 1

if acertada:
    print("Contraseña correcta")
else:
    print("Te has quedado sin intentos")
```

## Ejercicio 4

Dada `l = [3, 8, 1, 9, 4]`, recórrela con `for i in range(len(l))` y muestra la posición y el valor de cada elemento.

```python
l = [3, 8, 1, 9, 4]

for i in range(len(l)):
    print(i, l[i])
```
