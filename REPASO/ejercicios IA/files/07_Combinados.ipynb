# Ejercicios combinados (estilo examen)

Escribe tu solución en el bloque de código de cada ejercicio.

## Ejercicio 1

Pide al usuario una lista de notas separadas por comas (ej.: `7, 4.5, 9`), conviértelas a números (controlando errores con `try/except`) y muestra la media, la nota mayor y cuántas aprueban (>= 5).

## Ejercicio 2

Crea una función que reciba un diccionario `{producto: precio}` y un nombre de producto, y devuelva el precio. Si no existe, que devuelva un mensaje de error usando `try/except`.

## Ejercicio 3

Lee un fichero de texto, guarda cada línea en una lista y cuenta cuántas líneas tienen más de 20 caracteres. Controla `FileNotFoundError`.

## Ejercicio 4

Dada una lista de tuplas `[("Ane", 8), ("Jon", 4), ("Leire", 6)]` (nombre, nota), crea una función que devuelva un diccionario solo con los aprobados y otra que devuelva el nombre del alumno con mejor nota.

## Ejercicio 5

Pide nombres de alumnos y sus notas en un bucle hasta que el usuario escriba `fin`. Guarda los datos en un diccionario (controlando que la nota sea un número) y al final muestra la media y quién tiene la nota más alta.
