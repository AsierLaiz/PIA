### 2. Bi zerrenda indizearen arabera kateatu
Idatzi programa bat bi zerrenda indizearen arabera batzeko. Sortu zerrenda berri bat jatorrizko zerrendaren 0 indizean dagoen elementua duena, gero 1 indizean dagoen elementua, eta horrela azken elementura arte. Geratzen diren elementuak zerrenda berriaren amaieran gehituko dira.

Erabili `zip()` funtzioa

list1 = ["M", "na", "i", "Ke"]
list2 = ["y", "me", "s", "lly"]

Emaitza: 
['My', 'name', 'is', 'Kelly']

```python
list1 = ["M", "na", "i", "Ke"] 
list2 = ["y", "me", "s", "lly"]
# La función zip() toma las dos listas y empareja los elementos de las mismas posiciones, creando pares de tuplas con los elementos correspondientes de list1 y list2.
list3 = [i + j for i, j in zip(list1, list2)]
print(list3)

#opción 2

list1 = ["M", "na", "i", "Ke"]
list2 = ["y", "me", "s", "lly"]

# Crear una lista vacía para almacenar los resultados
list3 = []

# Usar un bucle for con un rango basado en la longitud de las listas
for i in range(len(list1)):
    # Concatenar los elementos correspondientes de list1 y list2
    list3.append(list1[i] + list2[i])

print(list3)
```

### 3. Zerrenda bateko elementu bakoitza bere karratu bihurtu
Zenbaki zerrenda bat emanda, idatzi programa bat zerrendako elementu bakoitza bere karratu bihurtzeko.

numbers = [1, 2, 3, 4, 5, 6, 7]

Emaitza: [1, 4, 9, 16, 25, 36, 49]

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
# result list
res = []
for i in numbers:
    # calculate square and add to the result list
    res.append(i * i)
print(res)
```

### 4. Kateatu bi zerrenda hurrengo ordenan

list1 = ["Hello ", "take "]
list2 = ["Dear", "Sir"]

Emaitza: ['Hello Dear', 'Hello Sir', 'take Dear', 'take Sir']

```python
list1 = ["Hello ", "take "]
list2 = ["Dear", "Sir"]

res = [x + y for x in list1 for y in list2]
print(res)
```

### 5. Bi zerrendak aldi berean iteratu
Bi Python zerrenda emanda, idatzi programa bat bi zerrendak aldi berean iteratzeko eta list1-eko elementuak jatorrizko ordenan eta list2-eko elementuak alderantzizko ordenan bistaratzeko.

list1 = [10, 20, 30, 40]

Emaitza: 

10 400

20 300

30 200

40 100

### 6. Kendu kate hutsak kateen zerrendatik

```python
list1 = ["Mike", "", "Emma", "Kelly", "", "Brad"]

# remove None from list1 and convert result into list
res = list(filter(None, list1))
print(res)
```

### 7. Gehitu elementu berri bat zerrendara elementu espezifiko baten ondoren

Idatzi Python programa bat hurrengo zerrendako 6000. elementuaren ondoren 7000. elementua gehitzeko

list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]

Emaitza: [10, 20, [300, 400, [5000, 6000, 7000], 500], 30, 40]

```python
list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]

# understand indexing
# list1[0] = 10
# list1[1] = 20
# list1[2] = [300, 400, [5000, 6000], 500]
# list1[2][2] = [5000, 6000]
# list1[2][2][1] = 6000

# solution
list1[2][2].append(7000)
print(list1)
```

### 8. Zerrenda txertatu bat luzatzea azpizerrenda bat gehituz

Zerrenda txertatu bat emanda, idatzi programa bat hura luzatzeko ["h", "i", "j"] azpizerrenda gehituz, hurrengo zerrendaren itxura izan dezan.

list1 = ["a", "b", ["c", ["d", "e", ["f", "g"], "k"], "l"], "m", "n"]

Gehitzeko azpizerrenda: sub_list = ["h", "i", "j"]

Emaitza: ['a', 'b', ['c', ['d', 'e', ['f', 'g', 'h', 'i', 'j'], 'k'], 'l'], 'm', 'n']

```python
list1 = ["a", "b", ["c", ["d", "e", ["f", "g"], "k"], "l"], "m", "n"]
sub_list = ["h", "i", "j"]

# understand indexing
# list1[2] = ['c', ['d', 'e', ['f', 'g'], 'k'], 'l']
# list1[2][1] = ['d', 'e', ['f', 'g'], 'k']
# list1[2][1][2] = ['f', 'g']

# solution
list1[2][1][2].extend(sub_list)
print(list1)
```

### 9. Ordeztu zerrenda-elementua balio berri batekin, aurkitzen bada

Python zerrenda bat emanda, idatzi programa bat zerrendan 20 balioa aurkitzeko eta, baldin badago, 200-rekin ordezkatzeko. Eguneratu elementuaren lehen agerpena bakarrik.

list1 = [5, 10, 15, 20, 25, 50, 20]

Emaitza: [5, 10, 15, 200, 25, 50, 20]


```python
list1 = [5, 10, 15, 20, 25, 50, 20]

# get the first occurrence index
index = list1.index(20)

# update item present at location
list1[index] = 200
print(list1)
```

### 10. Elementu espezifiko baten agerpen guztiak zerrenda batetik kendu.

Python zerrenda bat emanda, idatzi programa bat 20 elementuaren agerpen guztiak kentzeko.

list1 = [5, 20, 15, 20, 25, 50, 20]

Emaitza: [5, 15, 25, 50]

```python

# Solución 1

list1 = [5, 20, 15, 20, 25, 50, 20]

while 20 in list1:
    list1.remove(20)
print(list1)

# Solución 2

list1 = [5, 20, 15, 20, 25, 50, 20]

# list comprehension
# remove specific items and return a new list
def remove_value(sample_list, val):
    return [i for i in sample_list if i != val]

res = remove_value(list1, 20)
print(res)
```
